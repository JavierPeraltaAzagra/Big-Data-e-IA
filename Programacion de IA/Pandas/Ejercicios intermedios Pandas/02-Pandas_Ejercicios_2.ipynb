## Ejercicio 1
Crea un DataFrame de 20x6 con los siguientes tipos de datos:
- Columnas numéricas con enteros y flotantes.
- Columna de fechas aleatorias en 2025.
- Columna categórica con valores como 'A', 'B', 'C'.
- Columna booleana aleatoria.
Muestra las primeras 5 filas y la estructura del DataFrame (`info()`).

```python
import pandas as pd
import numpy as np

col_int = np.random.randint(0, 100, 20)
col_float = np.random.uniform(0, 100, 20)
# Fechas aleatorias en 2025
fechas = pd.to_datetime(
    np.random.randint(
        # Se divide entre 10**9 para convertir de nanosegundos a segundos
        pd.Timestamp("2025-01-01").value // 10**9,
        pd.Timestamp("2025-12-31").value // 10**9,
        20
    ),
    unit="s"
)

# Choice elige aleatoriamente entre las opciones dadas
categorias = np.random.choice(["A", "B", "C"], size=20)
booleanos = np.random.choice([True, False], size=20)

df = pd.DataFrame({
    "Entero": col_int,
    "Float": col_float,
    "Fecha": fechas,
    "Categoría": categorias,
    "Booleano": booleanos
})

print("Primeras 5 filas:")
print(df.head(), "\n")
print("Información del DataFrame:")
print(df.info())
```

## Ejercicio 2
Introduce valores nulos en aproximadamente el 15% de las celdas del DataFrame.
- Reemplaza los nulos en columnas numéricas con la **media**.
- Reemplaza los nulos en columnas categóricas con el **valor más frecuente**.
- Muestra el resultado y verifica que no quedan nulos.

```python
prob = [0.15, 0.85]
mask = np.random.choice([True, False], size=df.shape, p=prob)
# .mask() reemplaza los valores True con NaN
df = df.mask(mask)
print(f"Df con nulos 15%:\n{df}\n")
df["Entero"] = df["Entero"].fillna(df["Entero"].mean())
df["Float"] = df["Float"].fillna(df["Float"].mean())
df["Fecha"] = df["Fecha"].fillna(df["Fecha"].mean())
df["Categoría"] = df["Categoría"].fillna(df["Categoría"].mode()[0])
df["Booleano"] = df["Booleano"].fillna(df["Booleano"].mode()[0])
print(f"Df sin nulos:\n{df}")
print(f"Nulos restantes:\n{df.isnull().sum()}")
```

## Ejercicio 4
Crea un DataFrame de 100 filas con tres columnas numéricas.
- Calcula la **correlación condicional** entre las columnas solo para las filas donde la columna booleana sea True.
- Identifica la pareja de columnas con mayor correlación condicional.

```python
col_num1 = np.random.rand(100) * 100
col_num2 = np.random.rand(100) * 100
col_num3 = np.random.rand(100) * 100
col_bool = np.random.choice([True, False], size=100)
df = pd.DataFrame({
    "Num1": col_num1,
    "Num2": col_num2,
    "Num3": col_num3,
    "Booleano": col_bool
})
# Creo un dataframe únicamente con los valores booleanos True
df_true = df[df["Booleano"] == True]
# La correlación condicional se calcula con .corr(), una correlacion de 1 significa que si un número de una col
# sube una unidad, el de la otra columna también sube 1 unidad.
corr_cond = df_true[["Num1", "Num2", "Num3"]].corr()
print("Correlación condicional (True):")
print(corr_cond, "\n")
# Creo un DataFrame con valores absolutos (positivos) porque, para compararlos y ver la correlación más alta
# los valores tienen que estar con el mismo símbolo (+ o -). Ej: -0.15 tiene más correlación que 0.09
corr_abs = corr_cond.abs()
# Cambio los valores de la diagonal porque esos valores siempre tienen 1 de correlación, ya que se comparan
# con ellas mismas
np.fill_diagonal(corr_abs.values, np.nan)
print("Correlación condicional (absolutas y sin valores en la diagonal):")
print(corr_abs, "\n")
# .stack() sirve para pasar las columnas de un DataFrame a otro DataFrame distinto, donde se muestra el valor
# de las columnas por cada fila
print("Print de DataFrame con .stack()")
print(corr_abs.stack(), "\n")
# .idxmax() saca el nombre de las filas con el valor máximo
mejor_pareja = corr_abs.stack().idxmax()
valor_pareja = corr_cond.loc[mejor_pareja[0], mejor_pareja[1]]
print(mejor_pareja, valor_pareja)
```

## Ejercicio 5
Genera un DataFrame con ventas de 50 productos en 12 meses (columna 'Producto', columnas 'Mes_1' a 'Mes_12').
- Calcula la suma anual por producto.
- Crea una columna con la **media mensual** de cada producto.
- Ordena el DataFrame por total anual descendente.

```python
df = pd.DataFrame({
    "Producto": np.array(range(1,51)),
    "Mes_1": np.random.randint(0, 500, 50),
    "Mes_2": np.random.randint(0, 500, 50),
    "Mes_3": np.random.randint(0, 500, 50),
    "Mes_4": np.random.randint(0, 500, 50),
    "Mes_5": np.random.randint(0, 500, 50),
    "Mes_6": np.random.randint(0, 500, 50),
    "Mes_7": np.random.randint(0, 500, 50),
    "Mes_8": np.random.randint(0, 500, 50),
    "Mes_9": np.random.randint(0, 500, 50),
    "Mes_10": np.random.randint(0, 500, 50),
    "Mes_11": np.random.randint(0, 500, 50),
    "Mes_12": np.random.randint(0, 500, 50),
})
# Suma anual por producto
df["Total_Anual"] = df.loc[:, "Mes_1":"Mes_12"].sum(axis=1)
# Media mensual por producto
df["Media mensual"] = df.loc[:, "Mes_1":"Mes_12"].mean(axis=1)
# DataFrame ordenado por total anual descendente
df.sort_values("Total_Anual", ascending= False)
```

```python
# 33. Agrupamos por 'clase' y calculamos la media de 'ancho_petalo' para cada grupo
df_group = df.groupby("clase")["ancho_petalo"].mean()
```

```python
# 34. Asignamos un nombre a la Series resultante de la agrupación
df_group.name = "media_ancho_petalo"
```

```python
# 35. Mostramos la Series con la media del ancho del pétalo por clase
df_group
```

```python
# 36. Unimos la Series agrupada al DataFrame original usando join
df_join = df.join(df_group, on=["clase"], how="inner")
```

```python
# 37. Mostramos el DataFrame resultante tras la unión
df_join
```

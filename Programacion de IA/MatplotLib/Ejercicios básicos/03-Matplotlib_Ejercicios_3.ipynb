## Ejercicio 3

Muestra un gráfico de líneas con tres líneas de distintos colores y grosores. Cada una de ellas representará los valores de las funciones trigonométricas seno, coseno y tangente entre los 0 y los 720 grados. Además, muestra un título y las etiquetas (labels):

```python
grados = np.linspace(0, 720)
radianes = np.deg2rad(grados)

y_sin = np.sin(radianes)
y_cos = np.cos(radianes)
y_tan = np.tan(radianes)

plt.figure(figsize=(10, 6))
plt.plot(grados, y_sin, label="Seno", color="blue", linewidth=2)
plt.plot(grados, y_cos, label="Coseno", color="green", linewidth=1.5)
plt.plot(grados, y_tan, label="Tangente", color="red", linewidth=1)

plt.title("Funciones trigonométricas entre 0° y 720°")
plt.xlabel("Grados")
plt.ylabel("Valor")
plt.legend()
plt.show()
```

## Ejercicio 10

Hay una página web llamada [Kaggle](https://www.kaggle.com/datasets) donde se publican distintos datasets y trabajos de machine learning sobre los mismos. Descárgate un dataset de tu elección y muestra datos relevantes del mismo utilizando las gráficas vistas hasta ahora:

```python
df = pd.read_csv("Global_Climate_Change_Data_2020_2025.csv")

# Gráfico de barras: Top 10 países por emisiones totales de CO2
co2_por_pais = df.groupby("Country")["CO2_Emissions(Mt)"].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
co2_por_pais.plot(kind="bar", color="blue")
plt.title("Top 10 países por emisiones acumuladas de CO2 2020-2025")
plt.ylabel("Emisiones de CO2")
plt.xlabel("País")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Gráfico de pastel: Distribución de emisiones de CO2 por continente
co2_por_continente = df.groupby("Continent")["CO2_Emissions(Mt)"].sum()
plt.figure(figsize=(8,8))
colores = ["lightblue", "red", "orange", "green", "yellow", "purple"]
plt.pie(co2_por_continente, labels=co2_por_continente.index, autopct="%1.1f%%", startangle=90, colors=colores)
plt.title("Distribución de emisiones de CO2 por continente (2020-2025)")
plt.show()

# Gráfico scatter: temperatura media vs subida del nivel del mar
plt.figure(figsize=(10,6))
plt.scatter(df["Avg_Temperature(°C)"], df["Sea_Level_Rise(mm)"], alpha=0.7)
plt.xlabel("Avg Temperature (°C)")
plt.ylabel("Sea Level Rise (mm)")
plt.title("Temperatura media vs subida del nivel del mar")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Ejercicio 11

Sobre el mismo dataset anterior u otro nuevo, muestra tres gráficas distintas a las aprendidas hasta el momento. Para ello puedes consultar la web de la [documentación de Matplotlib](https://matplotlib.org/stable/gallery/index.html):

```python
co2_anyo_continente = df.groupby(['Year','Continent'])['CO2_Emissions(Mt)'].sum().unstack(fill_value=0).sort_index()

plt.figure(figsize=(12,6))
co2_anyo_continente.plot(kind='area', stacked=True, figsize=(12,6), cmap='tab20')
plt.title('Emisiones de CO2 apiladas por continente (2020-2025)')
plt.xlabel('Años')
plt.ylabel('Emisiones CO2')
plt.tight_layout()
plt.show()
```

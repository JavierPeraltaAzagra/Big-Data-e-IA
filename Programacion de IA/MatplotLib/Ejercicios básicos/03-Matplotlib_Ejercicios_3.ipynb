## Ejercicio 3

Muestra un gráfico de líneas con tres líneas de distintos colores y grosores. Cada una de ellas representará los valores de las funciones trigonométricas seno, coseno y tangente entre los 0 y los 720 grados. Además, muestra un título y las etiquetas (labels):

```python
grados = np.linspace(0, 720)
radianes = np.deg2rad(grados)

y_sin = np.sin(radianes)
y_cos = np.cos(radianes)
y_tan = np.tan(radianes)

plt.figure(figsize=(10, 6))
plt.plot(grados, y_sin, label="Seno", color="blue", linewidth=2)
plt.plot(grados, y_cos, label="Coseno", color="green", linewidth=1.5)
plt.plot(grados, y_tan, label="Tangente", color="red", linewidth=1)

plt.title("Funciones trigonométricas entre 0° y 720°")
plt.xlabel("Grados")
plt.ylabel("Valor")
plt.legend()
plt.show()
```

## Ejercicio 8

Muestra un gráfico circular estilizado con la distribución de tus horas del día:

```python
datos = [6,8,1,3,6]
actividades = ["Dormir", "Trabajar", "Labores domésticas", "Estudiar", "Otros"]
colores = ["lightblue", "red", "orange", "green", "yellow"]

plt.pie(datos, labels = actividades, colors = colores, startangle = 90,
        shadow = True, explode = (0.25,0,0,0,0), autopct = "%1.1f%%")

plt.title("Gráfico circular")
plt.show()
```

## Ejercicio 10

Hay una página web llamada [Kaggle](https://www.kaggle.com/datasets) donde se publican distintos datasets y trabajos de machine learning sobre los mismos. Descárgate un dataset de tu elección y muestra datos relevantes del mismo utilizando las gráficas vistas hasta ahora:

```python
df = pd.read_csv("Global_Climate_Change_Data_2020_2025.csv")

x = df["Continent"].unique()
y = df.groupby("Continent")[["Avg_Temperature(°C)"]].mean()

# configuramos los datos de cada matriz en el gráfico
plt.bar(x, y, label = "línea 1", color = "lightblue", width = 0.5)

# definimos el título y nombre de los ejes
plt.title("Gráfico de barras")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()

plt.show()

datos = df.groupby("Continent")[["Avg_Temperature(°C)"]].mean()
continentes = df["Continent"].unique()
colores = ["lightblue", "red", "orange", "green", "yellow", "lightgreen"]

print(datos.values.flatten())
plt.pie(datos.values.flatten(), labels = continentes, colors = colores, startangle = 90,
        shadow = True, autopct = "%1.1f%%")

plt.title("Gráfico circular")
plt.show()
```

## NIVEL 1: Conceptos Básicos
Tu misión: Completa el código donde veas # TODO rellenando el patrón vacío (r\"\")."
Comprueba si tu resultado coincide con el ESPERADO.

```python
import re

# TODO 1: Búsqueda literal ignorando mayúsculas/minúsculas. Busca la palabra "error".
texto_n1 = "El sistema detectó un error grave. El servidor colapsó por otro Error. ERROR solucionado."

patron_1 = r"error" #Completar
print("1. Literales:", re.findall(patron_1, texto_n1, re.IGNORECASE))
# ESPERADO: ['error', 'Error', 'ERROR']
```

```python
# TODO 2: Encuentra palabras de 4 letras que empiecen por 'c' y terminen en 'o'.
texto_ej2 = "El caso del cono que cayó en el cato y rodó como un cubo."
patron_2 = r"c.{2}o" #Completar, pista...(.)
print("2. ", re.findall(patron_2, texto_ej2))
# ESPERADO: ['caso', 'cono', 'cato', 'como', 'cubo']
```

```python
# TODO 3: Devuelve las líneas que EMPIEZAN por la palabra "Admin".
logs = ["Admin: Login OK", "User: Login Error", "Admin: Logout"]
patron_3 = r"^Admin" #Pista ^
print("3. ):", [linea for linea in logs if re.search(patron_3, linea)])
# ESPERADO: ['Admin: Login OK', 'Admin: Logout']
```

```python
# TODO 4: Ancla de fin ($). Devuelve los archivos que TERMINAN exactamente con ".py".
archivos = ["script.py", "main.pyc", "test.py", "py.txt"]
patron_4 = r".py$"
print("4:", [archivo for archivo in archivos if re.search(patron_4, archivo)])
# ESPERADO: ['script.py', 'test.py']
```

## NIVEL 2: Clases de Caracteres y Rangos

```python
texto_n2 = "Pedido 45A: 3 ratones, 12 teclados. Total: 150 euros."

# TODO 5: Buscar todos los números (dígitos aislados). Usa \d.
patron_5 = r"\d"
print("\n5. Dígitos:", re.findall(patron_5, texto_n2))
# ESPERADO: ['4', '5', '3', '1', '2', '1', '5', '0']
```

```python
# TODO 6: Rangos [ ]. Extrae todas las vocales (minúsculas).
palabra_ej6 = "Murcielago"
patron_6 = r"[aeiou]"
print("6. Vocales:", re.findall(patron_6, palabra_ej6))
# ESPERADO: ['u', 'i', 'e', 'a', 'o']
```

```python
# TODO 7: Rangos negados [^ ]. Extrae lo que NO sea una vocal minúscula ni un espacio.
palabra_ej7 = "Murcielago volador"
patron_7 = r"[^aeiou\s]"
print("7. Consonantes:", re.findall(patron_7, palabra_ej7))
# ESPERADO: ['M', 'r', 'c', 'l', 'g', 'v', 'l', 'd', 'r']
```

```python
# TODO 8: El espacio en blanco \s. Usa re.split para dividir por los espacios.
frase_ej8 = "Datos    separados por   muchos   espacios"
patron_8 = r"\s+"
print("8. Split espacios:", re.split(patron_8, frase_ej8))
# ESPERADO: ['Datos', 'separados', 'por', 'muchos', 'espacios']
```

## NIVEL 3: Cuantificadores y Pereza

```python
texto_n3 = "El año 1998 fue clave. IPs: 192.168.1.1 y CP 28001."
```

```python
# TODO 9: Repetición exacta {n}. Encuentra años de exactamente 4 dígitos (Usa \b para los bordes).
patron_9 = r"\b\d{4}\b"
print("\n9. Años:", re.findall(patron_9, texto_n3))
# ESPERADO: ['1998']
```

```python
# TODO 10: Rango {n,m}. Encuentra palabras de entre 4 y 5 letras.
patron_10 = r"\w{4,5}"
print("10. Rango long.:", re.findall(patron_10, "Un gato rojo salta la tapia."))
# ESPERADO: ['gato', 'rojo', 'salta', 'tapia']
```

```python
# TODO 11: Uno o Más (+). Encuentra los números completos de cualquier longitud.
patron_11 = r"\d+"
print("11. Números enteros:", re.findall(patron_11, texto_n3))
# ESPERADO: ['1998', '192', '168', '1', '1', '28001']
```

```python
# TODO 12: Modo Perezoso (?). Extrae lo que hay DENTRO de las comillas, sin llevarte las comillas.
html_falso = 'El atributo style="color: red" y class="btn"'
patron_12 = r"\"(.*?)\""
print("12. Perezoso:", re.findall(patron_12, html_falso))
# ESPERADO: ['color: red', 'btn']
```

## NIVEL 4: Agrupacion

```python
texto_n4 = "Me gustan los perros y los gatos. Email: fp@educacion.es."
```

```python
# TODO 13: Busca "perros" o "gatos". Alternancia (OR)
patron_13 = r"perros | gatos"
print("\n13. Alternancia:", re.findall(patron_13, texto_n4))
# ESPERADO: ['perros', 'gatos']
```

```python
# TODO 14: Grupos de Captura. De un email, extrae SOLO el dominio (después de la @).
emails = "maria@gmail.com, carlos@outlook.es"
patron_14 = r"@(\w+\.\w+)"
print("14. Dominios:", re.findall(patron_14, emails))
# ESPERADO: ['gmail.com', 'outlook.es']
```

```python
# TODO 15: Múltiples grupos. De una fecha dd/mm/yyyy, extrae (día, mes, año).
fechas = "Hoy es 15/04/2024."
patron_15 = r"(\d{2})/(.{2})/(\d{4})"
print("15. Fechas tuplas:", re.findall(patron_15, fechas))
# ESPERADO: [('15', '04', '2024')]
```

```python
# TODO 16: Sustitución (re.sub). Cambia los primeros 6 dígitos de un teléfono por asteriscos.
tlfs = "Contacto: 611222333"
patron_16 = r"\d{6}(\d{3})" # Captura los últimos 3 dígitos y usa r"******\1" en el reemplazo
print("16. Ocultar tlf:", re.sub(patron_16, r"******\1", tlfs))
# ESPERADO: Contacto: ******333
```

# NIVEL 5: Lookarounds

```python
# TODO 17: Extrae hashtags SIN la almohadilla (usa lookbehind positivo (?<=...) ).
tweet = "Aprendiendo #Python y #Regex"
# Coge lo que hay despues de un # y los caracteres que no sean espacios
patron_17 = r"(?<=#)\S+"
print("\n17. Hashtags:", re.findall(patron_17, tweet))
# ESPERADO: ['Python', 'Regex']
```

```python
# TODO 18: URLs válidas (empiezan por http:// o https://).
texto_web = "Visita https://google.com. No vayas a ftp://malware.org."
patron_18 = r"https?://\S+"
print("18. URLs:", re.findall(patron_18, texto_web))
# ESPERADO: ['https://google.com']
```

```python
# TODO 19: MAC Address (6 grupos de 2 hexadecimanles separados por :).
texto_red = "MAC válida: 00:1A:2B:3C:4D:5E. MAC rota: 99:ZZ:11:22:33:44"
patron_19 = r""
print("19. MAC Address:", re.findall(patron_19, texto_red))
# ESPERADO: ['00:1A:2B:3C:4D:5E']
```

```python
# TODO 20: Contraseña segura (Mín 6 chars, 1 mayús, 1 minús, 1 número usando Lookaheads).
passwords = ["admin", "Seguridad99"]
patron_20 = r""

print("20. Auditoría de Contraseñas:")
for pwd in passwords:
    if re.match(patron_20, pwd):
        print(f" [+] '{pwd}' CUMPLE")
    else:
        print(f" [-] '{pwd}' NO CUMPLE")
# ESPERADO:
# [-] 'admin' NO CUMPLE
# [+] 'Seguridad99' CUMPLE
```

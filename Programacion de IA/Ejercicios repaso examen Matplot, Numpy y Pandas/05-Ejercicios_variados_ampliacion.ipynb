## Normalización min-max de matriz NumPy

1 - Genera una matriz de 5 filas y 5 columnas cuyos valores sean enteros aleatorios comprendidos entre 10 y 500.

2 - Aplica la normalización min-max a toda la matriz, usando la fórmula:
$$
valor normalizado = (valor original - mínimo) / (máximo - mínimo)
$$
Esto convierte todos los valores de la matriz a un rango entre 0 y 1.

3 - Muestra la matriz normalizada, redondeando los valores a 2 decimales para que sea más legible.

```python
#1
matriz = np.random.randint(10, 501, [5, 5])

#2
val_min = np.min(matriz)
val_max = np.max(matriz)
matriz_norm = (matriz - val_min) / (val_max - val_min)

#3
matriz_norm_red = np.round(matriz_norm, 2)
matriz_norm_red
```

## Cálculo de diferencias porcentuales

1 - Crea un DataFrame con dos columnas: ventas_2023 y ventas_2024, cada una con 12 valores aleatorios que representen las ventas de cada mes.

2 - Calcula la variación porcentual mensual entre las ventas de ambos años usando la fórmula:

$$
variación = ((ventas en 2024 - ventas en 2023) / ventas en 2023) * 100
$$

3 - Añade esta variación como una nueva columna en el DataFrame.

4 - Dibuja un gráfico de línea que muestre la evolución de la variación a lo largo de los 12 meses, utilizando marcadores en cada punto para mayor claridad.

```python
#1
datos = {"ventas_2023": np.random.randint(0, 100, 12),
         "ventas_2024": np.random.randint(0, 100, 12)}
df = pd.DataFrame(datos)
#2
variacion = ((df["ventas_2024"] - df["ventas_2023"])/df["ventas_2023"]) * 100
#3
df["variacion"] = variacion
#4
plt.plot(df.index ,df["variacion"])
plt.show()
```

## Normalización min-max de matriz NumPy

1 - Genera una matriz de 5 filas y 5 columnas cuyos valores sean enteros aleatorios comprendidos entre 10 y 500.

2 - Aplica la normalización min-max a toda la matriz, usando la fórmula:
$$
valor normalizado = (valor original - mínimo) / (máximo - mínimo)
$$
Esto convierte todos los valores de la matriz a un rango entre 0 y 1.

3 - Muestra la matriz normalizada, redondeando los valores a 2 decimales para que sea más legible.

```python
#1
matriz = np.random.randint(10, 501, [5, 5])

#2
val_min = np.min(matriz)
val_max = np.max(matriz)
matriz_norm = (matriz - val_min) / (val_max - val_min)

#3
matriz_norm_red = np.round(matriz_norm, 2)
matriz_norm_red
```

## Media móvil en una serie temporal sencilla

1 - Genera una serie temporal de 50 valores aleatorios siguiendo una distribución normal, representando datos con ruido.

2 - Calcula la media móvil de ventana 5 para suavizar la serie y captar tendencias locales.

**Nota:** una media móvil es un cálculo que suaviza los datos de una serie temporal tomando el promedio de un subconjunto de valores (una "ventana") que se desplaza a lo largo de la serie. Esto ayuda a identificar tendencias al reducir el impacto de las fluctuaciones de corto plazo. Se implementa fácilmente en pandas usando el método .rolling() seguido de la función de agregación como .mean(). 

3 - Crea un gráfico de línea que muestre tanto la serie original como la serie suavizada con la media móvil.

4 - Añade una leyenda para identificar claramente ambas curvas en el gráfico.

```python
#1
serie = pd.Series(np.random.normal(size=50))

#2
media_movil = serie.rolling(window=5).mean()

#3 y 4
plt.figure(figsize=(8,4))
plt.plot(serie, marker='o', markersize=4, linestyle='-', label='Serie original')
plt.plot(media_movil, color='red', linewidth=2, label='Media móvil (w=5)')
plt.title('Serie temporal y media móvil (ventana=5)')
plt.xlabel('Índice')
plt.ylabel('Valor')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## Eliminación de valores extremos (outliers)

1 - Genera un DataFrame con una columna que contiene datos aleatorios y algunos valores extremos añadidos manualmente.

2 - Calcula la media y la desviación estándar de la columna.

3 - Filtra los valores que se encuentran dentro de 3 desviaciones estándar respecto a la media, eliminando los outliers.

**Pista:**

- Calcula la media de la columna que quieres filtrar (datos).

- Calcula la desviación estándar de esa misma columna.

- Recorre los valores de la columna y comprueba, para cada valor, si su diferencia absoluta respecto a la media es menor o igual a 3 veces la desviación estándar.

- Conserva solo los valores que cumplen esa condición y elimina los demás.

4 - Muestra cuántos valores fueron eliminados y el DataFrame filtrado resultante.

```python
#1
base = np.random.normal(size=40)
outliers = np.array([200.0, -50.0, 160.0])
datos = np.concatenate([base, outliers])
df = pd.DataFrame({"datos": datos})

#2
media = df["datos"].mean()
desv = df["datos"].std()

#3
mask = df["datos"].sub(media).abs() <= 3 * desv
df_filtrado = df[mask].reset_index(drop=True)

#4
eliminados = len(df) - len(df_filtrado)
print(f"Totales: {len(df)}  |  Eliminados: {eliminados}")
df_filtrado
```

## Series de diferencias acumuladas (investigación)

1 - Genera una serie Pandas de 30 valores enteros aleatorios entre 1 y 9.

2 - Calcula las diferencias entre elementos consecutivos de la serie usando diff().

3 - Obtén la suma acumulada de esas diferencias para generar una nueva serie.

4 - Muestra la serie resultante con las diferencias acumuladas.

```python
#1
serie = pd.Series(np.random.randint(1, 10, 30), name="serie")

#2
diferencias = serie.diff()

#3
diferencias_acum = diferencias.cumsum()
diferencias_acum = pd.Series(diferencias_acum.values, name="diferencias_acum")

#4
print("Serie original:\n", serie.to_list())
print("\nDiferencias:\n", diferencias.to_list())
print("\nSuma acumulada de diferencias:\n", diferencias_acum.to_list())
diferencias_acum
```

## Generación de una matriz aleatoria sin repetidos (investigación)

1 - Genera una matriz 4×4 cuyos elementos sean los números del 1 al 16 en un orden aleatorio.

2 - Utiliza np.random.permutation para asegurar que los números no se repitan en la matriz.

3 - Comprueba si hay elementos repetidos comparando la longitud de la matriz aplanada con la longitud de los elementos únicos. Investiga la función flatten.

4 - Muestra la matriz generada y si existen o no valores repetidos.

```python
#1 y 2
perm = np.random.permutation(np.arange(1, 17))
matriz = perm.reshape(4, 4)

#3
flattened = matriz.flatten()
unicos = np.unique(flattened)
repetidos = len(flattened) != len(unicos)

#4
print("Matriz 4x4 aleatoria (valores 1..16):\n", matriz)
print("Valores repetidos:", repetidos)
```

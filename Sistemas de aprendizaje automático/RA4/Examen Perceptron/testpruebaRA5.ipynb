```python
# 2. (3P) En la siguiente versión de la clase perceptrón, realiza las siguientes tareas:
#  a. (0,5P) Cambia los bucles for por while para iterar por épocas y por cada época.
#  b. (1P) Modifica la clase Perceptron(...) para que devuelva también el número de épocas necesarias para entrenar el Perceptrón. Para ello,
#     incluye la variable 'num_epocas_' dentro del constructor de la clase Perceptron y, después, colócala dentro de la clase en los lugares adecuados
#     para que guarde el número de épocas en los que el Percetrón termina su entrenamiento. Finalmente, instancia un objeto de la clase
#     Perceptron, accede a este atributo del objeto y muestra su valor.
#  c. (1P) Para cada valor de tasa de aprendizaje en
#     tasas_aprendizaje = [0.1, 0.09, 0.08, 0.07, 0.06, 0.05, 0.04, 0.03, 0.02, 0.01, 0.009, 0.008, 0.007, 0.006, 0.005, 0.004, 0.003, 0.002, 0.001],
#     itera sobre un objeto Perceptron y, en cada iteración, guarda el número de épocas de entrenamiento de ese objeto Perceptron en la
#     lista numero_epocas = []. Muestra ambas listas.
#  d. (0.5P) Mediante una gráfica en Python y las listas anteriores, muestra como varía el número de épocas necesarias para el entrenamiento
#     de un objeto Perceptron cuando la tasa de aprendizaje cambia. Utiliza la función plot(..) para dibujar la gráfica, con el número de épocas
#     en el eje Y y las tasas de aprendizaje en el eje X. Escribe un título para la gráfica, etiqueta para el eje x y etiqueta para el eje y.
```

```python
# Versión interfaz estilo scikit-learn 
import random           # Permite generar números aleatorios
class Perceptron:
    def __init__(self, epocas, learning_rate=0.1):
        '''
        Constructor de la clase
        Se ejecuta automaticamente al instanciar un objeto de la clase
        e incializa las variables que se encuentran dentro del constructor.         
        '''
        self.epocas=epocas   # numero máximo de epocas de entrenamiento
        self.learning_rate = learning_rate # tasa de aprendizaje
        # El que variables en el constructor tengan valor None significa que se les
        # asignará valor después de inicializar (instanciar) el objeto
        # En scikit-learn se la convención guion bajo al final para un atributo aprendido
        # atributos aprendidos
        self.coef_ = None        # Vector de pesos del modelo, por ejemplo [w1, w2, w3, .....]
        self.intercept_ = None   # Bias o término independiente del modelo
        self.num_epocas_ = None
        # fin del constructor-------------------------------------------------------------------------

    def predict(self, x1, x2):
        # Predice la clase de un punto con coordenada x1 y x2
        
        # Extraemos los pesos almacenados en coef_
        # Necesita w1, w2 y b, pero no como argumentos, porque ya están guardados dentro del objeto (self.w1, self.w2 y self.b)
        w1 = self.coef_[0]        
        w2 = self.coef_[1]
        b  = self.intercept_
        # calculamos el valor de agregacion z
        # z = w1*x1 + w2*x2 + b
        z = w1 * x1 + w2 * x2 + b

        # predicción de la clase a la que pertenece el punto (x1, x2)
        if z >= 0:  # si z >= 0 el preceptrón predice clase 1
            return 1
        else:       # si z < 0 el preceptrón predice clase 1
            return 0

    def fit(self, X, y):
        # Función que entrena el perceptrón utilizando los datos
        # X: variables predictoras, generalmente correspondientes a X_train
        # y: etiquetas reales de las variables predictoras del conjunto de entrenamiento

        # Número de ejemplos del dataset (número de filas)
        m = len(X)

        # Número de características por ejemplo (por línea del dataset)
        n = len(X[0]) # número de columnas

        print("Filas del conjunto X:", m)
        print("Columnas del conjunto X:", n)

        # Inicializamos aleatoriamente los pesos del vector coef_
        # Habrá un peso por cada característica de entrada
        self.coef_ = [random.uniform(-0.5, 0.5) for _ in range(n)]

        # Inicializamos aleatoriamente el bias o término independiente
        self.intercept_ = random.uniform(-0.5, 0.5)
        
        # mostramos valores inciales de los pesos
        print('Valor inicial de los parámetros')
        print(f'coef_ = {self.coef_}  intercept_ = {self.intercept_}')

        # Bucle que controla el entrenamiento por épocas
        self.num_epocas_ = 0
        while self.num_epocas_ < self.epocas:
            errores_en_epoca = 0 # contador de errores por época
            # Bucle que recorre cada punto del dataset (un punto = una línea) por cada época
            k = 0
            while k < m:
                # extraemos las características de cada ejemplo

                x1 = X[k][0]     # característica 1 (columna 1) del punto de la fila k 
                x2 = X[k][1]     # característica 2 (columna 2) del punto de la fila k

                y_real = y[k]
                # realizamos una predicción del perceptrón para el punto de la fila k
                y_pred = self.predict(x1, x2)

                #error que comete el perceptrón en la predicción
                error = y_real - y_pred

                #Llevamos la cuenta de los errores cometidos en esta época
                if error !=0:
                    errores_en_epoca +=1

                # actualizamos los parámetros del perceptrón
                self.coef_[0] = self.coef_[0]  + self.learning_rate * error * x1
                self.coef_[1] = self.coef_[1]  + self.learning_rate * error * x2
                self.intercept_  = self.intercept_  + self.learning_rate * error
                k = k + 1

            # Errores cometidos en esta época
            print(f'Epoca: {self.num_epocas_ + 1} | Errores: {errores_en_epoca}')

            # Si no hay errores en esta época, terminamos el entrenamiento
            if errores_en_epoca == 0:
                print(f'Entrenamiento terminado después de {self.num_epocas_ + 1} epocas')
                # mostramos valores inciales de los pesos
                print('Valor final de los parámetros')
                print(f'coef_ = {self.coef_}  intercept_ = {self.intercept_}')
                return self # devolvemos el objeto ya entrenado
            self.num_epocas_ = self.num_epocas_ + 1
  
        # Si se alcanza el número máximo de épocas sin llegar a error cero
        print('Entrenamiento finalizado en el límite de épocas')
        print('Valor final de los parámetros')
        print(f'coef_ = {self.coef_}   intercept_ = {self.intercept_}')
        return self
X, y = make_blobs(
    n_samples=[40, 40],
    centers=[(2, 2), (7, 3)],
    random_state=42
)
tasas_aprendizaje = [0.1, 0.09, 0.08, 0.07, 0.06, 0.05, 0.04, 0.03, 0.02, 0.01, 0.009, 0.008, 0.007, 0.006, 0.005, 0.004, 0.003, 0.002, 0.001]
numero_epocas = []
for tasa in tasas_aprendizaje:
    clf = Perceptron(epocas=20, learning_rate=tasa)
    clf.fit(X, y)
    numero_epocas.append(clf.num_epocas_)
print("Tasas de aprendizaje usadas: ", tasas_aprendizaje)
print("Num épocas por tasa: ", numero_epocas)

plt.figure(figsize=(8, 5))
plt.plot(tasas_aprendizaje, numero_epocas, marker="o", linestyle="-")
plt.title("Variación del número de épocas según la tasa de aprendizaje")
plt.xlabel("Tasa de aprendizaje")
plt.ylabel("Número de épocas")
plt.grid(True)
plt.show()
```

```python
# 3. (3,5P) Modifica la clase la clase Perceptrón para generalizarla al manejo de cualquier número de dimensiones.
#    En este caso, no debemos utilizar las variables individuales de un dataset: x1, x2, ....., debemos utilizar vectores
#    de características. Sugerencia: se trata de ampliar la última clase Perceptron que hemos visto en el cuaderno
#    'Límite de visualización del Perceptrón' y usar la función np.dot(vector1, vector 2). Tened en cuenta, además, que la fila k de X
#    se obtiene de la forma x = X[k] -aquí están incluidos x = (x1, x2)-
# Versión interfaz estilo scikit-learn 
import random           # Permite generar números aleatorios
import numpy as np
class Perceptron:
    def __init__(self, epocas, learning_rate=0.1):
        '''
        Constructor de la clase
        Se ejecuta automaticamente al instanciar un objeto de la clase
        e incializa las variables que se encuentran dentro del constructor.         
        '''
        self.epocas=epocas   # numero máximo de epocas de entrenamiento
        self.learning_rate = learning_rate # tasa de aprendizaje
        # El que variables en el constructor tengan valor None significa que se les
        # asignará valor después de inicializar (instanciar) el objeto
        # En scikit-learn se la convención guion bajo al final para un atributo aprendido
        # atributos aprendidos
        self.coef_ = None        # Vector de pesos del modelo, por ejemplo [w1, w2, w3, .....]
        self.intercept_ = None   # Bias o término independiente del modelo
        self.num_epocas_ = None
        # fin del constructor-------------------------------------------------------------------------

    def predict(self, x):
        w = self.coef_
        b = self.intercept_
        z = np.dot(x, w) + b
        return 1 if z >= 0 else 0

    def fit(self, X, y):
        # Función que entrena el perceptrón utilizando los datos
        # X: variables predictoras, generalmente correspondientes a X_train
        # y: etiquetas reales de las variables predictoras del conjunto de entrenamiento

        # Número de ejemplos del dataset (número de filas)
        m = len(X)

        # Número de características por ejemplo (por línea del dataset)
        n = len(X[0]) # número de columnas

        print("Filas del conjunto X:", m)
        print("Columnas del conjunto X:", n)

        # Inicializamos aleatoriamente los pesos del vector coef_
        # Habrá un peso por cada característica de entrada
        self.coef_ = [random.uniform(-0.5, 0.5) for _ in range(n)]

        # Inicializamos aleatoriamente el bias o término independiente
        self.intercept_ = random.uniform(-0.5, 0.5)
        
        # mostramos valores inciales de los pesos
        print('Valor inicial de los parámetros')
        print(f'coef_ = {self.coef_}  intercept_ = {self.intercept_}')

        # Bucle que controla el entrenamiento por épocas
        self.num_epocas_ = 0
        while self.num_epocas_ < self.epocas:
            errores_en_epoca = 0 # contador de errores por época
            # Bucle que recorre cada punto del dataset (un punto = una línea) por cada época
            k = 0
            while k < m:
                # extraemos las características de cada ejemplo
                x = X[k]
                y_real = y[k]
                # realizamos una predicción del perceptrón para el punto de la fila k
                y_pred = self.predict(x)

                #error que comete el perceptrón en la predicción
                error = y_real - y_pred

                #Llevamos la cuenta de los errores cometidos en esta época
                if error !=0:
                    errores_en_epoca +=1

                # actualizamos los parámetros del perceptrón
                self.coef_ = self.coef_ + np.dot(self.learning_rate * error, x)
                self.intercept_  = self.intercept_  + self.learning_rate * error
                k = k + 1

            # Errores cometidos en esta época
            print(f'Epoca: {self.num_epocas_ + 1} | Errores: {errores_en_epoca}')

            # Si no hay errores en esta época, terminamos el entrenamiento
            if errores_en_epoca == 0:
                print(f'Entrenamiento terminado después de {self.num_epocas_ + 1} epocas')
                # mostramos valores inciales de los pesos
                print('Valor final de los parámetros')
                print(f'coef_ = {self.coef_}  intercept_ = {self.intercept_}')
                return self # devolvemos el objeto ya entrenado
            self.num_epocas_ = self.num_epocas_ + 1
  
        # Si se alcanza el número máximo de épocas sin llegar a error cero
        print('Entrenamiento finalizado en el límite de épocas')
        print('Valor final de los parámetros')
        print(f'coef_ = {self.coef_}   intercept_ = {self.intercept_}')
        return self
X, y = make_blobs(
    n_samples=[40, 40],
    centers=[(2, 2), (7, 3)],
    random_state=42
)
tasas_aprendizaje = [0.1, 0.09, 0.08, 0.07, 0.06, 0.05, 0.04, 0.03, 0.02, 0.01, 0.009, 0.008, 0.007, 0.006, 0.005, 0.004, 0.003, 0.002, 0.001]
numero_epocas = []
for tasa in tasas_aprendizaje:
    clf = Perceptron(epocas=20, learning_rate=tasa)
    clf.fit(X, y)
    numero_epocas.append(clf.num_epocas_)
print("Tasas de aprendizaje usadas: ", tasas_aprendizaje)
print("Num épocas por tasa: ", numero_epocas)

plt.figure(figsize=(8, 5))
plt.plot(tasas_aprendizaje, numero_epocas, marker="o", linestyle="-")
plt.title("Variación del número de épocas según la tasa de aprendizaje")
plt.xlabel("Tasa de aprendizaje")
plt.ylabel("Número de épocas")
plt.grid(True)
plt.show()
```

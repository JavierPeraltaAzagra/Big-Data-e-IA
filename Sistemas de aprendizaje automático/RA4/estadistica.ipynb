Media(μ-mu)- Es el valor medio de la distribución.

Desviación estandar(σ-sigma) – Representa la dispersión de la distribución alrededor de la media. 

````python
# Función de densidad de probabilidad para una distribución normal
def funcion_densidad_probabilidad(x, media, desviacion):
    return np.exp((-(x - media)**2) / (2 * desviacion**2)) / (desviacion * np.sqrt(2 * np.pi))
````

````python
#Gráfico de la función de distribución normal con matplotlib
plt.figure(figsize=(20,10))
x = np.arange(-4,15,0.1) # x = [-4, -3.9, -3.8, ........ , 3.9, 4]
media = 0
desviacion = 1
y = funcion_densidad_probabilidad(x, media, desviacion)
plt.plot(x,y, label='media = 0, desviación típica = 1')

media = 5
desviacion = 3
y = funcion_densidad_probabilidad(x, media, desviacion)
plt.plot(x,y, label='media = 5, desviación típica = 3')
plt.title('FUNCIONES DE DENSIDAD DE PROBABILIDAD NORMAL DE LA VARIABLE ALETAORIA X')
plt.xlabel('valores de x')
plt.ylabel('probabilidad de x')
plt.legend()
plt.show()
````

````python
import seaborn as sns  # Importa seaborn para gráficos estadísticos
import numpy as np
import matplotlib.pyplot as plt

# Parámetros de la primera distribución normal
mu = 0              # Media de la distribución normal.
sigma = 1                  # Desviación estándar (típica) de la distribución.


y = np.random.normal(mu, sigma, 20000) # Genera 20000 valores aleatorios siguiendo una normal N(0,1)
plt.figure(figsize=(20,10)) # Crea una figura grande de 20x10 pulgadas
plt.xticks(range(-10, 20, 1)) # Define las marcas del eje X desde -10 hasta 19

# Dibuja el histograma normalizado (densidad) con curva KDE. Incluye label para la leyenda
sns.histplot(y,        # Datos a graficar (muestras de la distribución)
             kde=True, # Dibuja una curva suave de densidad que aproxima la distribución de los datos, complementando el histograma.
             stat='density', # Normaliza el histograma para que represente densidad de probabilidad en lugar de frecuencias.
             label='N(0, 1)'
            )

# Etiquetas de los ejes y título del gráfico
plt.xlabel('x')                                  # Etiqueta del eje X
plt.ylabel('y (Probabilidad de x)')              # Etiqueta del eje Y
plt.title("FUNCIÓN DE DENSIDAD DE PROBABILIDAD NORMAL DE LA VARIABLE ALEATORIA X")

# --- SEGUNDA DISTRIBUCIÓN NORMAL ---

# Nuevos parámetros para otra distribución normal
mu = 5              # media de la distribución
sigma = 3                  # desviación (estándar) típica de la distribución
y = np.random.normal(mu, sigma, 20000) # Genera 20000 valores de una distribución normal N(5,3)

# Dibuja el segundo histograma en la misma figura con otro label (etiqueta)
sns.histplot(y, kde=True, stat='density', label='N(5, 3)')
plt.legend() # Añade la leyenda al gráfico
plt.show() # Muestra el gráfico con ambas distribuciones
````

# Interpretación del Q–Q Plot de `df.income`

El gráfico Q–Q (Quantile–Quantile) compara los cuantiles observados de la variable `income` con los cuantiles teóricos de una distribución normal. La línea roja representa la distribución normal ideal: si los datos siguieran una distribución normal, los puntos caerían sobre esa línea.

---

## 1. Ajuste en la parte central
La zona central de los datos se aproxima a la línea roja, lo que indica que los valores medios tienen cierta similitud con una distribución normal.

---

## 2. Desviación clara en la cola superior
Los puntos correspondientes a valores altos de `income` se alejan fuertemente hacia arriba respecto a la línea roja.

Esto indica:
- Presencia de **valores extremos altos** (outliers).
- Una **cola derecha** típica de datos económicos (ingresos, riqueza).
- Que la distribución tiene **asimetría positiva**.

---

## 3. La cola inferior también muestra cierta desviación
Aunque menos marcada que la superior, los puntos de la cola izquierda también se separan de la línea roja, lo que indica que los valores bajos tampoco siguen perfectamente una distribución normal.

---

## Conclusión general
La variable **`df.income` NO sigue una distribución normal**.

Evidencias:
- Los puntos no forman una línea recta.
- La cola derecha se dispara claramente por encima de la línea teórica.
- El patrón de desviación es sistemático y no aleatorio.

Este comportamiento es **normal** en variables económicas como ingresos, que suelen presentar:
- Asimetría a la derecha,
- Cola larga,
- Pocos valores extremadamente altos (long tail distribution).

### Explicación de `fig, ax = plt.subplots(figsize=(20,10))`

La instrucción:

```python
fig, ax = plt.subplots(figsize=(20,10)
```

crea una figura y uno o más ejes (subplots) en Matplotlib.

Esta función crea:

1. **fig**  es el objeto Figura, que representa el lienzo o área completa del gráfico.

2. **ax**  es el objeto Axes, que representa el gráfico en sí (el sistema de ejes donde se dibuja).

En este caso devuelve un único eje, por eso se asigna a fig, ax.

````python
# Creamos la figura y los ejes explícitamente
fig, ax = plt.subplots(figsize=(20,10))

# Histograma KDE Peso_kg
sns.histplot(
    df.Peso_kg,
    kde=True,            # Añade la curva KDE (estimación suave de la densidad a partir de los datos)
    stat='density',      # Histograma en escala de densidad. El histograma se normaliza para representar densidad (área total ≈ 1)
    ax=ax                # Dibujamos en el eje ax
)

# Etiquetas y título del gráfico
ax.set_xlabel('Peso (kg)')
ax.set_ylabel('Densidad de probabilidad')
ax.set_title("HISTOGRAMA DE DENSIDAD DE PROBABILIDAD Y FUNCIÓN DE DENSIDAD DE LA VARIABLE ALEATORIA Peso (kg)")

# CÁLCULO DE LA función de distribución de probabilidad normal teórica
media = df.Peso_kg.mean()  # media muestral
s = df.Peso_kg.std() # desvi. típica muestral

x = np.arange(df.Peso_kg.min(), df.Peso_kg.max(), 1) # Rango de valores x

y = funcion_densidad_probabilidad(x, media, s) # Valores de la función de eensidad teórica normal

# Dibujamos la curva normal teórica en el mismo eje (ax) que el histograma KDE
ax.plot(
    x,   # valores de x
    y,   # valores de probabilidad (entre 0 y 1)
    color='red',
    label=f'fdp media {media:.2f}, desviación {s:.2f}'
)
# Línea vertical que representa la media en el mismo conjunto de ejes que el diagrama KDE y la curva  normal teórica anterior
ax.axvline(
    media,               # Posición en el eje X
    color="limegreen",   # Color de la línea
    linestyle="--",      # Línea discontinua
    linewidth=2,         # Grosor
    label=f'Media {media:.2f}'
)
ax.legend()  # mostramos la leyenda (label) en el conjunto de ejes ax
plt.show()   # visualizamos la figura
````

La función `st.norm.cdf((x, loc=mu, scale=sigma)` pertenece al módulo `scipy.stats` y se utiliza para calcular la **función de distribución acumulada (CDF)** de una distribución normal con media µ y desviación típica σ.

x: valor en el que se evalúa la probabilidad acumulada

loc: media 

μ de la distribución normal

scale: desviación típica 

σ de la distribución normal

Devuelve la probabilidad:

$$
P(X \le x)
$$

donde:

$$
X \sim N(\text{µ},\, \sigma)
$$

Es decir, la función devuelve la probabilidad de que un valor tomado al azar de esa distribución normal sea menor o igual que *x*.

# Explicación de la función `z_alfa = st.norm.ppf(alfa)`

La función `st.norm.ppf()` pertenece a *scipy.stats* y calcula el **percent point function (PPF)**, también llamado **cuantil** o **valor crítico** de la distribución normal estándar.

En el ejemplo:

```python
z_alfa = st.norm.ppf(alfa)
```
Se obtiene el **valor de z** tal que:

Consideramos una población de 10000 valores del peso de una población normalmente distribuida de parámetros:

media de la población (mu)(suponemos que es desconocida)

desviacion de la población (sigma) 7.89 (conocida)

Hemos generado 100 muestras aleatorias simples de tamaño 50 de dicha población y, ahora, vamos a generar un intervalo de confianza al 95% para el parámetro μ (media poblacional del peso) para cada muestra.


````python
#Nivel de significación
n = 50 #tamaño de la muestra
nc = 0.95 #nivel de confianza
alfa = 1 - nc #nivel de significacion
mediaMuestral = medias[0]

print(f'Media muestral: {mediaMuestral}')
print(f'Desviación típica poblacional: {sigma}')
print(f'Nivel de significación (alfa): {round(alfa,3)}')

#Cálculo del valor crítico. Determinar el peso de forma que el 1 -alfa/2 = 1- 0.05/2 = 0.975 (97.5%) del alumnado tenga un peso inferior a ese peso.

#Límite superior del intervalo de confianza

valor_critico = st.norm.ppf(0.975)

print(f'\nValor critico: {valor_critico}')
#Calculamos el valor real a partir del valor tipificado del peso
peso_sup = mediaMuestral + valor_critico*sigma/math.sqrt(n)

print(f"\nEl 97.5% del alumnado tiene un peso (real) inferior a z (1-alfa/2) {peso_sup}:")

#Límite inferior del intervalo de confianza


#Calculamos el valor real a partir del valor tipificado del peso
peso_inf = mediaMuestral - valor_critico*sigma/math.sqrt(n)

print(f"\nEl 2.5% del alumnado tiene un peso (real) inferior a (z alfa/2) {peso_inf}:")

#Intervalo de confianza
print(f'\nIntervalo de confianza para la media poblacional \nIC = ({peso_inf},{peso_sup})')
````

````python
def generaICN(x, sigma, alfa):  
    # Entrada:
    # x: array de tamaño 50. Representa una muestra que contiene 50 valores de peso (kg)
    # sigma: desviación típca de la población, que hemos supuesto conocida.
    # alfa: nivel de significación. Usualmente 0.05
    
    # Devuelve:
    # Límite inferior del Intervalo de Confianza
    # Límite superior del Intervalo de Confianza
    
    lim_sup = np.array(x).mean() + st.norm.ppf(1-alfa/2)*sigma/math.sqrt(len(np.array(x)))
    lim_inf = np.array(x).mean() - st.norm.ppf(1-alfa/2)*sigma/math.sqrt(len(np.array(x)))
    print(f'\nIntervalo de confianza para la media poblacional \nIC = ({lim_inf},{lim_sup})')
    return lim_inf, lim_sup
````

## **Distribución t-student** ##

La distribución t, también conocida como distribución t de Student, es una distribución de probabilidad que se utiliza en estadística inferencial cuando el tamaño de la muestra es pequeño y se desconoce la desviación estándar de la población. Es una variación de la distribución normal con colas más pesadas, lo que la hace más apropiada para estimar la media de una población cuando el tamaño de la muestra es pequeño o cuando existe incertidumbre sobre la desviación estándar de la población.

La distribución t se caracteriza por sus grados de libertad (gl), que también determinan su forma. Los grados de libertad representan el número de características independientes en el conjunto de datos. A medida que aumentan los grados de libertad, la distribución t se aproxima a la forma de una distribución normal estándar.

Características de la distribución t

1. Simetría: La distribución t es simétrica respecto a su media.

2. Ubicación y escala: La distribución t generalmente tiene una media cero; sin embargo, su desviación estándar es mayor que cero debido a su cola más pesada.

3. Colas: La distribución t tiene colas más gruesas (colas más grandes), lo que significa que hay menos puntos cerca de la media en comparación con la distribución normal.

4. Forma: La forma de la distribución t depende de sus grados de libertad. Además, a medida que aumentan los grados de libertad, la distribución t se acerca más a una distribución normal.

5. El parámetros de la distribución t grados de libertad (df), se calcula como el tamaño de la muestra menos uno.

**¿Cuándo usar la distribución t?**

La distribución t de Student se utiliza cuando:

1. El tamaño de la muestra es 30 o menor.
   
2. Se desconoce la desviación estándar de la población (σ).

**¿En qué se diferencia la distribución t de la distribución normal?**

1. Se parece a una distribución normal, pero tiene colas más gruesas.
   
2. A medida que aumenta el tamaño de la muestra, la distribución t se aproxima gradualmente a la distribución normal.
   
3. Esto tiene en cuenta la incertidumbre adicional que se introduce al estimar la desviación estándar de la población a partir de una muestra pequeña.

````python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import t

x = np.linspace(-5, 5, 100)
gradosLibertad = [1, 2, 5, 10] # Grados de libertad


for df in gradosLibertad:
	y = t.pdf(x, df) # Using default location and scale parameters (0 and 1)
	plt.plot(x, y, label=f"Grados de libertad = {df}")

plt.xlabel('Desviaciones típicas')
plt.ylabel('Densidad de probabilidad')
plt.title('Funciones de densidad de probabilidad\n para diferentes grados de libertad')
plt.legend()
plt.show()
````

```python
import numpy as np
import matplotlib.pyplot as plt

def dibujar_recta(w1, w2, b):    
    #Dibuja la recta w1*x1 + w2*x2 + b = 0
    
    plt.figure(figsize=(5,5))
    # Creamos valores de x1 (eje horizontal)    np.linspace genera puntos entre -10 y 10
    x1 = np.linspace(-10, 10, 100)
    # Despejamos y de la ecuación:   w1*x1 + w2*x2 + b = 0  ->  x2 = -(w1*x1 + b)/w2 y  w2 != 0
    x2 = -(w1/w2) * x1 - (b/w2)

    # Dibujamos la recta usando matplotlib
    plt.plot(x1, x2, color="red")
    # Dibujamos los ejes coordenados
    plt.axhline(0)   # eje x1
    plt.axvline(0)   # eje x2

    # Leyendas
    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.title(f'Recta: ${w1}x_1 + {w2}x_2 + ({b}) = 0$')
    #plt.title(r"Gráfica recta $w_1 x_1 + w_2 x_2 + b = 0$")
    plt.xlim(-2, 5) # definimos rango del eje x
    plt.ylim(-2, 7) # definimos rango del eje y
    plt.grid(True)
    # Mostramos la figura en pantalla
    plt.show()
```


$$
\textbf{Fórmula pendiente de una recta y punto de corte con el eje } x_2 \textbf{ a partir de la fórmula general de la recta}
$$


Partimos de la recta

$$
w_1 x_1 + w_2 x_2 + b = 0
$$

(suponiendo $w_2 \neq 0$).

---

### Forma explícita

Despejando $x_2$:

$$
w_2 x_2 = -w_1 x_1 - b
$$

$$
x_2 = -\frac{w_1}{w_2}x_1 - \frac{b}{w_2}
$$



---

### Pendiente

$$
m = -\frac{w_1}{w_2}
$$

---

### Punto de corte con el eje $x_2$

El corte con el eje $x_2$ ocurre cuando $x_1 = 0$:

$$
x_2 = -\frac{b}{w_2}
$$

Por tanto, el punto de corte es:

$$
(0,\,-\frac{b}{w_2})
$$

---

### Resumen

$$
x_2 = \underbrace{-\frac{w_1}{w_2}}_{\text{pendiente}}\,x_1 \;+\; \underbrace{\left(-\frac{b}{w_2}\right)}_{\text{corte en }x_2}
$$

**Caso especial:** si $w_2 = 0$, la recta es vertical y no se puede expresar como $x_2 = m x_1 + c$.

```python
# Efecto de b en la gráfica de la recta

def dibujar_rectas(w1, w2, rango_b):
    """
    Dibuja la recta w1*x1 + w2*x2 + b = 0
    """
    plt.figure(figsize=(8, 8))
    # Creamos valores de x1 (eje horizontal)    np.linspace genera puntos entre -10 y 10
    x1 = np.linspace(-10, 10, 200)

    # Despejamos y de la ecuación:   w1*x1 + w2*x2 + b = 0  ->  x2 = -(w1*x1 + b)/w2 y  w2 != 0
    for b in rango_b:
        x2 = -(w1 * x1 + b) / w2
        # Dibujamos la recta usando matplotlib
        plt.plot(x1, x2, label=f'b = {b}')
        
    # Dibujamos los ejes coordenados
    plt.axhline(0)   # eje x1
    plt.axvline(0)   # eje x2

    # Leyendas
    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.title(
        f'Rectas: ${w1}x_1 + {w2}x_2 + b = 0$'
    )
    plt.grid(True) 
    plt.legend()
    # Mostramos la figura en pantalla
    plt.show()

rango_b = [b for b in range (-12, 13, 3)]
rango_b
dibujar_rectas(w1 = 4, w2 = 1, rango_b = rango_b)
```

# Perceptrón binario (clases 0/1) — 1 época con los datos de clase

## Parámetros iniciales

Recta inicial (según la figura):

$$
1x_1 + (-2)x_2 + (1)=0
\Rightarrow w_1=1,\; w_2=-2,\; b=1
$$

Tasa de aprendizaje (**alfa**):

$$
\alpha = 0.1 \quad \text{(tasa de aprendizaje)}
$$



## Estructura del perceptrón:

### 1) Agregación (**función de agregación**)

$$
z = w_1x_1 + w_2x_2 + b
$$

### 2) Salida del perceptrón (**función de activación**) 

$$
a =
\begin{cases}
1 & \text{si } z \ge 0\\
0 & \text{si } z < 0
\end{cases}
$$

### 3) Error: es la diferencia entre la etiqueta del grupo asignado a un punto por el perceptrón (a) y la que tiene realmente (y).
#### Si un punto pertenece al grupo 1 (y=1) y el perceptrón le asigna 0 (a=0), el error es e = y - a = 1 - 0 = 0.
#### Si un punto pertenece al grupo 0 (y = 0) y el perceptrón le asigna 0 (a = 0), el error es e = y - a = 0 - 1 = -1. 

$$
e = (y-a)
$$

### 4) Actualización de los parámetros del Perceptrón: w1, w2 y b. De esta manera, el Perceptrón encuentra ('aprende') los parámetros óptimos que le permitan dividir correctamente los grupos de puntos sus clases correspondientes, de forma automática. No necesitamos encontrar estos parámetros de forma manual.

$$
\Large
w_1 \leftarrow w_1 + \alpha\, e\, x_1
$$

$$
\Large
w_2 \leftarrow w_2 + \alpha\, e\, x_2
$$

$$
\Large
b \leftarrow b + \alpha\, e
$$

# Época 1 (cálculos paso a paso) con $\alpha=0.1$

## Punto 1: $x=(0.8,0.8),\; y=0$

**Agregación:**

$$
z = 1(0.8) + (-2)(0.8) + 1 = 0.8 - 1.6 + 1 = 0.2
$$

**Activación:**

$$
z=0.2\ge 0 \Rightarrow a=1
$$

**Error:**

$$
e = y-a = 0-1 = -1
$$

**Actualización:**

$$
w_1 \leftarrow 1 + 0.1(-1)(0.8) = 1 - 0.08 = 0.92
$$

$$
w_2 \leftarrow -2 + 0.1(-1)(0.8) = -2 - 0.08 = -2.08
$$

$$
b \leftarrow 1 + 0.1(-1) = 1 - 0.1 = 0.9
$$

**Estado tras el punto 1:**

$$
w_1=0.92,\; w_2=-2.08,\; b=0.9
$$

```python
import numpy as np

# Funcion escalon binaria (0/1)
def activacion(z):
    # Si z >= 0 -> 1, si z < 0 -> 0
    return 1 if z >= 0 else 0

# Calcula z = w1*x1 + w2*x2 + b (formula explicita)
def agregacion(w1, w2, b, x1, x2):
    # z = w1*x1 + w2*x2 + b
    return w1*x1 + w2*x2 + b

# Ejecuta 1 epoca completa e imprime todas las operaciones
def perceptron_una_epoca(X, y, w1, w2, b, alfa=0.1):
    # Contador de actualizaciones (errores)
    actualizaciones = 0

    print("EJEMPLO DE APRENDIZAJE (1 EPOCA)")
    
    print("----------------------------------------")
    print(f"Inicial: w1={w1}, w2={w2}, b={b}, alfa={alfa}")

    # Recorremos los puntos en el orden dado
    for i in range(len(X)):
        x1, x2 = X[i]
        yi = y[i]

        # 1) Calculo de z
        z = agregacion(w1, w2, b, x1, x2)

        # 2) Prediccion a
        a = activacion(z)

        # 3) Error e
        e = yi - a

        # Mostrar calculos paso a paso
        print("\n----------------------------------------")
        print(f"Punto {i+1}: x=({x1}, {x2}), y={yi}")
        print(f"  z = w1*x1 + w2*x2 + b")
        print(f"    = ({w1})*({x1}) + ({w2})*({x2}) + ({b})")
        print(f"    = {z:.4f}")
        print(f"  a = activacion(z) = {a}  (porque z {'>=' if z>=0 else '<'} 0)")
        print(f"  e = (y - a) = {yi} - {a} = {e}")

        # 4) Actualizacion si hay error
        if e != 0:
            w1_new = w1 + alfa*e*x1
            w2_new = w2 + alfa*e*x2
            b_new  = b  + alfa*e

            print("  ACTUALIZACION (porque error e es diferente 0):")
            print(f"    w1 <- w1 + eta*e*x1 = {w1} + {alfa}*({e})*({x1}) = {w1_new}")
            print(f"    w2 <- w2 + eta*e*x2 = {w2} + {alfa}*({e})*({x2}) = {w2_new}")
            print(f"    b  <- b  + eta*e     = {b} + {alfa}*({e}) = {b_new}")

            w1, w2, b = w1_new, w2_new, b_new
            actualizaciones += 1
        else:
            print("  Sin actualizacion (e = 0).")

        print(f"  Estado tras el punto {i+1}: w1={w1}, w2={w2}, b={b}")

    print("\n========================================")
    print(f"Fin de la epoca: actualizaciones (errores) = {actualizaciones}")
    print(f"Parametros finales: w1={w1}, w2={w2}, b={b}")
    return w1, w2, b

# ------------------------------
# DATOS DE CLASE
# ------------------------------

# Grupo 0
x0 = [2, 3, 3]
y0 = [3, 3, 2]

# Grupo 1
x1 = [2, 1]
y1 = [1, 1]

# Construimos X (x1,x2) e y (0/1)
# Puntos del grupo 0
X0 = np.column_stack((x0, y0))
print(X0)
# Puntos del grupo 1
X1 = np.column_stack((x1, y1))
print(X1)
# Dataset completo
X = np.vstack((X0, X1))
print(X)
#y = np.array([0]*len(x0) + [1]*len(x1))
#print(y)
# Parametros iniciales (los de tu figura)
w1_init = 0.5
w2_init = -0.3
b_init  = 1
y = [1,1,1,0,0]

# Ejecutamos 1 epoca (eta=1 para cuentas faciles)
w1_final, w2_final, b_final = perceptron_una_epoca(X, y, w1_init, w2_init, b_init, alfa=0.1)
```

# Perceptrón simple — Pseudocódigo

El objetivo del **perceptrón simple** es la clasificación binaria (clases 0 y 1) mediante la construcción de una línea recta que separa dos grupos de puntos.



## 1. Variables principales

- $X$: conjunto de entrenamiento. Cada ejemplo es un vector: $(x_1, x_2)$

- $y$: etiquetas reales del conjunto (lista). Cada etiqueta es $0$ o $1$

- $(w_1, w_2)$: vector de pesos

- $b$: sesgo (*bias*)

- $\alpha\$: tasa de aprendizaje (`learning_rate`)

- `epocas`: número de veces que se recorre el conjunto de entrenamiento

## 2. Idea matemática del perceptrón

Para un ejemplo $(x_1, x_2)$, se calcula:

$$
\Large
z = w_1x_1 + w_2x_2 + b
$$

La salida predicha $\hat{y}$ se obtiene con una función escalón:

$$
\Large
\hat{y} = a(z) =
\begin{cases}
1 & \text{si } z \ge 0 \\
0 & \text{si } z < 0
\end{cases}
$$

Se calcula el error:

$$
\Large
\text{error} = y - \hat{y}
$$

Y se actualizan los parámetros:

$$
\Large
w_1 \leftarrow w_1 + \alpha\, error\, x_1
$$

$$
\Large
w_2 \leftarrow w_2 + \alpha\, error\, x_2
$$

$$
\Large
b \leftarrow b + \alpha\, error
$$

```python
#-----------------------------------------------------#
# Versión simplificada del algoritmo del Perceptrón   #
#-----------------------------------------------------#


import random  # Para generar números aleatorios (pesos y bias iniciales)
import numpy as np

def activacion(z):
    # Función escalón: convierte un valor real z en una clase (0 o 1)
    # Si z >= 0 => clase 1; si z < 0 => clase 0
    if z >= 0:
        return 1
    else:
        return 0

def calcular_z(x1, x2, w1, w2, b):    
    # Calcula el valor lineal del perceptrón antes de activar:    
    return   w1 * x1 + w2 * x2 + b # Devolvemos el valor de z

def predecir(x1, x2, w1, w2, b):
    # Predice la clase (0 o 1) para un punto x
    z = calcular_z(x1, x2, w1, w2, b)  # Calculamos el valor lineal
    y_pred = activacion(z)             # Aplicamos activación para obtener 0 o 1
    return y_pred                      # Devolvemos la clase predicha

def entrenar_perceptron(X, y, epocas=100, learning_rate=0.1):
    # Entrena el perceptrón con:
    # X: lista de ejemplos (puntos) (cada ejemplo es una lista de características)
    # y: lista de etiquetas reales (0 o 1)
    # epocas: máximo de pasadas por el dataset
    # learning_rate: tamaño del paso de actualización (alpha)   

   
    m = len(X) # Número de puntos en la tabla (filas de X)
    n = len(X[0]) # Número de características, en este caso x1 y x2 son dos características por cada línea de X

    print("Filas:", m)
    print("Columnas:", n)

    
    # Inicialización aleatoria de w y b (valores pequeños).  Pesos iniciales    
    w1 = random.uniform(-0.5, 0.5)
    w2 = random.uniform(-0.5, 0.5)
    b = random.uniform(-0.5, 0.5)
    
    print("Pesos iniciales w:", w1, w2)
    print("Bias inicial b:", b)
    print("-" * 50)
    
    # Entrenamiento por épocas
    # ------------------------------------------------------------
    for epoca in range(epocas):
        errores_en_epoca = 0  # Contador de ejemplos mal clasificados en esta época

        # Recorremos los ejemplos (puntos-líneas de la tabla)
        for k in range(m):
            x1 = X[k][0]    # primer valor de la fila k de la columna 0 (la columna de los x1)
            x2 = X[k][1]    # segundo valor de la fila k de la columna 0 (la columna de los x1)
            y_real = y[k]   # Etiqueta real del ejemplo k
            
            y_pred = predecir(x1, x2, w1, w2, b) # Predicción actual (en esta época)
            
            error = y_real - y_pred              # Error (y real - y predicha)

            # Contamos si hubo error
            if error != 0:
                errores_en_epoca = errores_en_epoca + 1

            # Actualización de pesos
            w1 = w1 + learning_rate * error * x1 # Actualización de peso w1
            w2 = w2 + learning_rate * error * x2 # Actualización de peso w2 
            b = b + learning_rate * error        # Actualización del bias b

        # Resumen por época
        print(f"Época {epoca:03d} | Errores: {errores_en_epoca}")
        
        # Criterio de parada        
        if errores_en_epoca == 0:
            print("Algoritmo de aprendizaje terminado por 0 errores en la época.")
            break

    return w1, w2, b  # Devolvemos los parámetros aprendidos de w1, w2 y b
#
```

```python
# Dividir en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(
    X,                # Matriz de características de entrada (features)-atributos de entrada-variables predictoras
    y,                # Vector/serie con la variable objetivo (labels-eneste caso: class)
    test_size=0.2,    # Proporción para el conjunto de prueba y bel conjunto de entrenamiento: 20% test, 80% train
    random_state=42,  # Semilla para que la división sea reproducible
    shuffle=True,      # Barajar aleatoriamente antes de dividir (recomendado si no hay una serie temporal)
    stratify=y         # mantiene proporciones de clases
)
```

```python
# Entrenamos el perceptrón
w1, w2, b = entrenar_perceptron(X_train, y_train, epocas=300, learning_rate=0.1)

print("-" * 50)
print(f'w1 = {w1}  w2 = {w2}  b = {b}') # valores finales de w1, w2, y b
print("-" * 50)
```

```python
# Probamos predicciones finales (en el conjunto de test) y mostramos z
# Probamos predicciones finales para los puntos (x1, x2) del conjunto de test
y_pred = []
for x1, x2 in X_test:
    y_pred.append(predecir(x1, x2, w1, w2, b))
    print(f'({x1},{x2}) | predicción={predecir(x1, x2, w1, w2, b)}')
y_pred
```

```python
# Matriz de confusión con conteos (para KNN)
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score
import matplotlib.pyplot as plt

labels = sorted(set(y_train))   # clases del problema
# Matriz de confusión (solo conteos)
mc = confusion_matrix(y_test, y_pred, labels=labels)

# Crear lienzo para dibujar la matriz de confusión
fig, axes = plt.subplots(figsize=(8, 8))  # aumenta el tamaño: ancho x alto


# Visualización Matriz de Confusión. 

labels = sorted(set(y_train))   # clases del problema
disp = ConfusionMatrixDisplay(                            # disp es un objeto de tipo ConfusionMatrixDisplay
                             confusion_matrix=mc,         # Matriz de confusión calculada previamente (con confusion_matrix())
                             display_labels=labels  # Etiquetas de las clases que el modelo KNN reconoce (por ejemplo, [0, 1])
                             )
disp.plot(
         cmap="Blues",                                    # Define el mapa de colores usado para pintar las celdas de la matriz.
         ax=axes,                                         # Especifica en qué figura o subgráfico se debe dibujar la matriz.
         colorbar=False                                   # Indica si se muestra o no la barra lateral de color (que representa la escala de valores).
         )

# Personalizar título y aspecto
axes.set_title("Matriz de Confusión (KNN)", fontsize=14)
axes.set_xlabel("Clase predicha", fontsize=12)
axes.set_ylabel("Clase real", fontsize=12)
plt.tight_layout()
plt.show()
```

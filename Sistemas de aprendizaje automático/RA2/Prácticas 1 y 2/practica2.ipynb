# Interpretación general de un gráfico 3D
En un gráfico 3D se muestran tres variables numéricas de un conjunto de datos al mismo tiempo.  
Cada punto representa una observación (una fila del DataFrame), y su color indica la clase a la que pertenece.

## Interpretación visual por clases

1. **Agrupamiento por color**  
   Si los puntos de una misma clase se agrupan en una región concreta del espacio 3D, significa que esas tres variables tienen una relación característica que distingue a esa clase.  
   Cuanto más compacto y separado esté ese grupo, más claramente la variable de clase se diferencia de las demás.

2. **Separación entre clases**  
   Si los grupos de colores están bien separados en el espacio tridimensional, las tres variables utilizadas son buenas para discriminar entre clases.  
   Si, en cambio, las nubes de puntos se superponen, las clases son más difíciles de distinguir usando solo esas variables.

3. **Forma de los grupos**  
   - Un grupo alargado puede indicar correlación entre dos de las variables.  
   - Un grupo más disperso sugiere que esa clase tiene mayor variabilidad en esas dimensiones.  
   - Si los grupos se cruzan o mezclan, es señal de que la frontera entre clases no es lineal o no puede definirse fácilmente.

4. **Densidad de puntos**  
   La concentración de puntos en una región indica dónde se acumulan los valores más frecuentes.  
   Regiones más vacías pueden corresponder a combinaciones poco comunes o imposibles de las variables.

## Conclusión

El gráfico 3D permite visualizar simultáneamente tres variables y analizar la relación entre ellas para diferentes clases.  
Si las clases aparecen bien separadas, esas variables podrían servir como buenos predictores en un modelo de clasificación.  
Si las clases se solapan o presentan fronteras difusas, puede ser necesario incluir más variables o aplicar técnicas no lineales para lograr una separación efectiva.





# Interpretación del gráfico 3D del dataset Pima Indians Diabetes

El gráfico 3D muestra la relación entre tres variables biométricas relevantes en el estudio de la diabetes:

- **Eje X:** Glucosa en plasma (`plas`)  
- **Eje Y:** Índice de masa corporal (`mass`)  
- **Eje Z:** Edad (`age`)  

Cada punto del gráfico representa una persona del conjunto de datos.  
El color de los puntos diferencia las clases:
- **Clase 0:** No diabético  
- **Clase 1:** Diabético  

## 1. Gráfico general

El gráfico permite observar cómo se comportan los individuos en función de sus niveles de glucosa, masa corporal y edad.  
A simple vista, los puntos de ambas clases pueden solaparse parcialmente, pero tienden a agruparse en regiones distintas:

- Los individuos **diabéticos (Clase 1)** tienden a ocupar la región con **valores altos de glucosa y mayor masa corporal**, y con frecuencia son también **de mayor edad**.

## 2. Eje X: Glucosa en plasma (plas)

Esta variable muestra la diferencia más clara entre las clases.  
La clase 1 (diabéticos) presenta valores notablemente más altos de glucosa, mientras que la clase 0 se concentra en niveles normales o bajos.  
Por tanto, **plas es la variable más discriminante del gráfico**: separa visualmente a los grupos a lo largo del eje X.

## 3. Eje Y: Índice de masa corporal (mass)

El IMC también contribuye a la separación de clases.  
Las personas con diabetes tienden a tener **mayor masa corporal**, lo que desplaza sus puntos hacia valores altos en el eje Y. No obstante, hay cierta superposición, ya que existen individuos con IMC alto sin diabetes y viceversa.  
Esto indica que el IMC, aunque útil, **no basta por sí solo** para diferenciar las clases.

## 4. Eje Z: Edad (age)

En general, los individuos diabéticos son **de mayor edad promedio**.  
Sin embargo, la variable `age` muestra **una menor capacidad de discriminación** en comparación con `plas` y `mass`, ya que existen personas jóvenes y mayores en ambas clases.

- En el espacio tridimensional formado por `plas`, `mass` y `age`, los grupos de clases **no están completamente separados**, pero sí muestran una tendencia:  
  los puntos de la **clase 1** se agrupan hacia valores altos de las tres variables.  
- El **solapamiento** en la zona de valores intermedios refleja la naturaleza multifactorial de la diabetes: la enfermedad no depende de un solo parámetro, sino de la combinación de varios factores.

## 6. Conclusión general

El gráfico 3D del dataset Pima Indians Diabetes muestra que:

- **Glucosa (plas)** es la variable más importante para distinguir entre personas diabéticas y no diabéticas.

# Actividad: Visualización 3D con diferentes orientaciones

## Objetivo

Explorar la distribución de los datos de un conjunto (`DataFrame`) en tres dimensiones, usando tres variables numéricas.  
El objetivo es observar cómo **la orientación del gráfico 3D afecta la interpretación visual** de la separación entre clases o la relación entre variables.

## Descripción de la Actividad

1. Define una función que reciba como entrada:
   - Un `DataFrame`
   - Los nombres de **tres variables numéricas** (una por eje)
   - Dos valores numéricos (`elev` y `azim`) para definir la orientación

2. La función deberá generar un **gráfico 3D** con los datos del dataset, diferenciando las clases por color (asumiendo que la variable de clase se llama `'class'`).

3. Finalmente, llama a la función **seis veces** con diferentes orientaciones para mostrar seis vistas distintas del mismo gráfico.


````python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
def grafico_3d(df, x_var, y_var, z_var, elev, azim):
    """
    Dibuja un gráfico 3D de tres variables numéricas con una orientación definida.
    Parámetros:
    df    : DataFrame con los datos
    x_var : nombre de la variable para el eje X
    y_var : nombre de la variable para el eje Y
    z_var : nombre de la variable para el eje Z
    elev  : ángulo de elevación de la cámara
    azim  : ángulo de rotación (azimut)
    """
    plt.figure(figsize=(8, 6))
    ax = plt.axes(projection='3d')
    plt.title(f'Gráfico 3D: {x_var}, {y_var}, {z_var} (elev={elev}, azim={azim})')
    # Clase 0
    x0 = df[df['class'] == 0][x_var]
    y0 = df[df['class'] == 0][y_var]
    z0 = df[df['class'] == 0][z_var]
    ax.scatter3D(x0, y0, z0, color='skyblue', alpha=0.7, label='Clase 0')
    # Clase 1
    x1 = df[df['class'] == 1][x_var]
    y1 = df[df['class'] == 1][y_var]
    z1 = df[df['class'] == 1][z_var]
    ax.scatter3D(x1, y1, z1, color='salmon', alpha=0.7, label='Clase 1')

    ax.set_xlabel(x_var)
    ax.set_ylabel(y_var)
    ax.set_zlabel(z_var)
    ax.legend()
    # Cambiar orientación
    ax.view_init(elev=elev, azim=azim)
    plt.tight_layout()
    plt.show()
````

````python
# Gráfico 1
grafico_3d(df, 'plas', 'mass', 'age', elev=20, azim=30)

# Gráfico 2
grafico_3d(df, 'plas', 'mass', 'age', elev=30, azim=60)

# Gráfico 3
grafico_3d(df, 'plas', 'mass', 'age', elev=45, azim=90)

# Gráfico 4
grafico_3d(df, 'plas', 'mass', 'age', elev=60, azim=120)

# Gráfico 5
grafico_3d(df, 'plas', 'mass', 'age', elev=15, azim=180)

# Gráfico 6
grafico_3d(df, 'plas', 'mass', 'age', elev=30, azim=270)
````

### **División del dataset en conjunto de prueba y conjunto de entrenamiento** ###

````python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Separar variables predictoras y objetivo
X = df.drop(columns=['class'])
y = df['class']


# Dividir en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(
    X,                # Matriz de características de entrada (features)-atributos de entrada-variables predictoras
    y,                # Vector/serie con la variable objetivo (labels-eneste caso: class)
    test_size=0.2,    # Proporción para el conjunto de prueba y bel conjunto de entrenamiento: 20% test, 80% train
    random_state=42,  # Semilla para que la división sea reproducible
    shuffle=True      # Barajar aleatoriamente antes de dividir (recomendado si no hay una serie temporal)
)

````

### Barajado (shuffle)

Si shuffle=True, los datos se mezclan aleatoriamente antes de separarlos.

Si shuffle=False, los datos se dividen en el orden en que están en el DataFrame (por ejemplo, las primeras filas para entrenamiento y las últimas para test).

En la mayoría de los datasets, el orden de las filas podría no es aleatorio. Por ejemplo, las primeras filas podrían pertenecer a una sola clase y las últimas a otra. Si no se baraja (shuffle), podríamos obtener conjuntos de entrenamiento y prueba desequilibrados o sesgados, lo que afectaría el rendimiento del modelo y la validez de la evaluación.

NO usar shuffle=True cuando los datos tienen una secuencia temporal (series de tiempo o datos cronológicos). En ese caso, no se debe mezclar, porque el conjunto de prueba debe representar datos futuros, no aleatorios.

En Matplotlib, cuando se crea un gráfico de barras (barplot, countplot, plt.bar()…), cada grupo de barras se agrupa en un objeto llamado BarContainer.

Un BarContainer es un objeto de Matplotlib que almacena todas las barras (rectángulos) que pertenecen a un mismo grupo o categoría dentro de un gráfico de barras. Por ejemplo, si el gráfico muestra tres grupos (Original, Entrenamiento y Prueba), cada grupo genera su propio objeto BarContainer.

Un "artist" en Matplotlib es cualquier elemento dibujado en la figura: una barra, una línea, un texto, un eje, etc. En este caso, “2 artists” significa que cada BarContainer tiene dos barras (una por cada clase, 0 y 1).

# Escalado de variables en el dataset Pima Indians Diabetes

## 1. Contexto del dataset

El conjunto de datos *Pima Indians Diabetes* contiene información médica de mujeres de origen Pima. Su objetivo es predecir si una persona tiene o no diabetes (variable `class` = 0 o 1) a partir de ocho variables numéricas:

| Variable | Descripción | Tipo de dato |
|-----------|--------------|--------------|
| preg | Número de embarazos | Entero |
| plas | Glucosa en plasma | Numérica continua |
| pres | Presión arterial diastólica | Numérica continua |
| skin | Grosor del pliegue cutáneo tricipital | Numérica continua |
| insu | Insulina sérica (μU/mL) | Numérica continua |
| mass | Índice de masa corporal (IMC) | Numérica continua |
| pedi | Función pediátrica de diabetes | Numérica continua |
| age  | Edad (años) | Entero |
| class | Variable objetivo (0 = no diabético, 1 = diabético) | Categórica binaria |


## 2. Necesidad del escalado

Las variables numéricas del dataset tienen **unidades y rangos muy diferentes**:

`plas` (glucosa) puede variar entre 50 y 200  
`insu` (insulina) puede llegar a valores superiores a 600  
`mass` (IMC) suele estar entre 15 y 50  
`pedi` (función pediátrica) toma valores pequeños (0.1–2.5)

Estos **diferentes órdenes de magnitud** afectan directamente a los algoritmos que calculan **distancias** (por ejemplo, KNN, SVM, PCA, K-Means) y se **regularizan** con penalizaciones L1/L2 (por ejemplo, regresión logística o redes neuronales).  

Sin escalado, las variables con valores grandes (como `insu`) **dominan el comportamiento del modelo**, reduciendo la importancia de otras variables más pequeñas (como `pedi`).

Por tanto, **es necesario escalar las variables numéricas** antes del entrenamiento.


## 3. Tipos de escalado aplicables

### a) **Normalización Min-Max**
Convierte los valores al rango [0, 1]:


x_escalado = (x - x_min)/(x_max - x_min)


**Ventajas:**
- Mantiene la forma de la distribución original.  
- Ideal para redes neuronales o métodos basados en distancia (KNN, SVM).

**Inconveniente:**
- Sensible a valores extremos (*outliers*).


### b) **Estandarización**
Centra los datos en 0 y escala la desviación estándar a 1:

x_escalado = (x - u)/desv_tipica


**Ventajas:**
- Menos afectado por las unidades originales.  
- Adecuado para la mayoría de los modelos lineales o de gradiente (Logistic Regression, SVM, PCA).  
- A menudo mejora la convergencia en modelos que usan optimización numérica.

**Inconveniente:**
- Los valores resultantes pueden ser negativos.  
- Puede ser menos intuitivo para interpretación directa.


**Data Leak**

El objeto StandardScaler calcula la media y desviación estándar usando todos los datos (entrenamiento + prueba). Es decir, el escalador (o cualquier transformador) usa estadísticas globales del dataset (como la media o varianza de cada columna). Si incluye los datos del conjunto de prueba en estos cálculos, entonces el modelo estará ajustado a información que debería ser desconocida. Por lo tanto, la información del conjunto de prueba tendrá influencia sobre el escalado del conjunto de entrenamiento.

**data leak: El modelo aprende de forma indirecta sobre el conjunto de prueba, antes de ser evaluado.**

Siempre se debe:

1. Dividir los datos primero,

2. Ajustar el escalador solo con el conjunto de entrenamiento,

3. Aplicar la misma transformación al conjunto de prueba.

### Reducción de la dimensionalidad

La reducción de la dimensionalidad es una técnica básica de preprocesado de datos, que consiste en la reducción del número de variables. Es un método estadístico que transforma las variables originales en un nuevo conjunto de variables, que son combinaciones lineales de las variables originales. Estas nuevas variables se denominan componentes principales.

Hay varias razones por las que puede ser interesante tratar de reducir el número de dimensiones del conjunto de datos:

- Se reduce la complejidad, con lo que el modelo y sus resultados se hacen más fácilmente interpretables.
  
- Por razones de ahorro, tanto en coste como en tiempo. Reduciendo las dimensiones, se mejora el rendimiento computacional.

- Eliminar redundancia, algunas variables están correlacionadas (ej. plas con test, o mass con skin).

- Reducir ruido, eliminar componentes poco informativos mejora la generalización.

- Visualizar datos: reducir a 2 o 3 dimensiones permite graficar la estructura del dataset.

En nuestro dataset, cada variable es una dimensión del espacio: (por ejemplo, edad, peso, glucosa, presión…). Los registros del dataset están distribuidos en ese espacio multidimensional.

PCA busca una nueva base de coordenadas (ejes nuevos) donde:

El primer eje (PC1) apunta en la dirección de mayor variabilidad de los datos.

El segundo eje (PC2) apunta en la segunda dirección más importante, y es perpendicular al primero.

Y así sucesivamente.

Podemos proyectar los datos en estos nuevos ejes y conservar casi toda la información con menos dimensiones (atributos del dataset).
  
**"maldición de la dimensionalidad”.**

Maldición de la dimensionalidad: A medida que aumenta el número de dimensiones, los puntos de datos se vuelven dispersos, lo que dificulta que los modelos de aprendizaje automático encuentren patrones.



````python
from sklearn.decomposition import PCA # Importa la clase PCA desde el módulo sklearn.decomposition.
pca = PCA(n_components=3) # Encuentra las 3 direcciones principales que capturan la mayor parte de la variabilidad de los datos.
pca # Muestra el objeto pca, que contiene la configuración inicial del modelo.

# Cada componente principal es una nueva variable (combinación lineal de las originales) 
# que resume la información del dataset en una dirección de máxima varianza. En este caso:

# PC1: dirección donde los datos varían más.
# PC2: segunda dirección de máxima variación, perpendicular a la primera.
# PC3: tercera dirección, también perpendicular a las anteriores.
````

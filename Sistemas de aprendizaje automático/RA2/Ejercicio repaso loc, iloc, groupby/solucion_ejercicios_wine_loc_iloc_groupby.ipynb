## Función para graficar y analizar una variable numérica

### **Objetivo**

Desarrollar una función en Python que permita representar gráficamente la distribución de una variable numérica de un conjunto de datos, utilizando un diagrama de caja (boxplot). Además, la función debe calcular y mostrar las principales estadísticas descriptivas de dicha variable.

### **Instrucciones**

1. **Crear la función**
   
Define una función llamada `boxplot_variable(df, var)` que reciba como parámetros:

- **`df`**: un *DataFrame* de pandas con los datos.
  
- **`var`**: una cadena (*str*) con el nombre de la columna numérica a analizar.



### **Dentro de la función**

2. Dibuja el boxplot de la variable seleccionada usando `sns.boxplot()` o `plt.boxplot()`.

   
   
3. Calcula las estadísticas principales y almacénalas en un diccionario vacío llamado **`stats`**:

   - **Q1** → primer cuartil: `df[var].quantile(0.25)`
   - **Q3** → tercer cuartil: `df[var].quantile(0.75)`
   - **median** → mediana: `df[var].median()`
   - **mean** → media: `df[var].mean()`
   - **min** → mínimo: `df[var].min()`
   - **max** → máximo: `df[var].max()`
   - **IQR** → rango intercuartílico: `Q3 - Q1`
   - **lim_inf** → límite inferior: `Q1 - 1.5 * IQR`
   - **lim_sup** → límite superior: `Q3 + 1.5 * IQR`
          

4. Agrega líneas horizontales en el gráfico que indiquen:

   
   - Media (`mean`)
   - Mediana (`median`)
   - Máximo (`max`)
   - Mínimo (`min`)

     
   Usa un bucle `for` con **`plt.axhline()`** para dibujarlas con diferentes colores y estilos.  Agrega textos al gráfico con los valores de las   estadísticas (Q1, Q3, límites, etc.). Usa otro bucle `for` con **`plt.text()`** para mostrar los valores.


5. Muestra el gráfico:

   
   - Agrega leyenda con `plt.legend()`.  
   - Activa la cuadrícula con `plt.grid(True)`.  
   - Muestra la figura con `plt.show()`.
   

6.  Retorna el diccionario `stats` con las estadísticas calculadas.



### **Ejemplo de uso**

````python
import matplotlib.pyplot as plt

def boxplot_variable(df, var):
    """
    Dibuja un diagrama de caja (boxplot) para una variable numérica y muestra en el gráfico las principales estadísticas descriptivas.

    Parámetros:
    -----------
    df : pandas.DataFrame        #Conjunto de datos.
    var : str                    #Nombre de la columna numérica a analizar.

    Retorna:
    --------
    stats : dict                 #Diccionario con las estadísticas calculadas.
    """

    # Gráfico boxplot 
    plt.figure(figsize=(10, 10))
    plt.boxplot(df[var])

    # Título y etiquetas 
    plt.title(f'Distribución de {var}', fontsize=14)
    plt.ylabel(var)

    # Cálculo de estadísticas 
    Q1 = df[var].quantile(0.25)  # Primer cuartil
    Q3 = df[var].quantile(0.75)  # Tercer cuartil
    median = df[var].median()    # Mediana
    mean = df[var].mean()        # Media
    max_val = df[var].max()      # Máximo
    min_val = df[var].min()      # Mínimo

    # Líneas de referencia
    plt.axhline(mean, color='red', linestyle='--', label=f'Media: {mean:.2f}')
    plt.axhline(median, color='green', linestyle='-.', label=f'Mediana: {median:.2f}')
    plt.axhline(max_val, color='orange', linestyle=':', label=f'Máximo: {max_val:.2f}')
    plt.axhline(min_val, color='blue', linestyle='--', label=f'Mínimo: {min_val:.2f}')

    # Leyendas de cuartiles
    plt.text(1.1, Q1, f'Q1 = {Q1:.1f}', color='purple')
    plt.text(1.1, Q3, f'Q3 = {Q3:.1f}', color='purple')

    # Leyendas de máximo y mínimo
    plt.text(1.1, max_val, f'Máx = {max_val:.1f}', color='black')
    plt.text(1.1, min_val, f'Mín = {min_val:.1f}', color='black')

    # Cálculo de límites superior e inferior
    IQR = Q3 - Q1
    lim_sup = Q3 + 1.5 * IQR
    lim_inf = Q1 - 1.5 * IQR

    # Textos de límites
    plt.text(0.8, lim_sup, f'Lím_sup = {lim_sup:.1f}', color='gray')
    plt.text(0.8, lim_inf, f'Lím_inf = {lim_inf:.1f}', color='gray')

    # Ajustes finales
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # Retornar diccionario con estadísticas
    stats = {
        'Q1': Q1,
        'Q3': Q3,
        'median': median,
        'mean': mean,
        'min': min_val,
        'max': max_val,
        'IQR': IQR,
        'lim_inf': lim_inf,
        'lim_sup': lim_sup
    }

    return stats
````

````python
# Versión más elaborada y compacta, recomendada para reutilizar y compactar el código.

import matplotlib.pyplot as plt

def boxplot_variable(df, var):
    """
    Representa un diagrama de caja (boxplot) para una variable numérica,
    y calcula sus principales estadísticas descriptivas.

    Parámetros
    ----------
    df : pandas.DataFrame        #Conjunto de datos.
    var : str                    #Nombre de la columna numérica a analizar.

    Retorna
    -------
    stats : dict
        Diccionario con las estadísticas calculadas.
    """
 
    # Dibujo del boxplot
    plt.figure(figsize=(6, 8))
    plt.boxplot(df[var],
                vert=True,
                patch_artist=True, # permite colorear la caja (por defecto es transparente).
                boxprops=dict(facecolor="lightblue", color="blue"), # color de la caja y su borde.
                medianprops=dict(color="orange", linewidth=2)) # color y grosor de la línea de la mediana
    
    plt.title(f"Diagrama de Caja de la variable '{var}'")
    plt.ylabel(var)

    # Cálculo de estadísticas
    stats = {}
    stats['Q1'] = df[var].quantile(0.25)
    stats['Q3'] = df[var].quantile(0.75)
    stats['median'] = df[var].median()
    stats['mean'] = df[var].mean()
    stats['min'] = df[var].min()
    stats['max'] = df[var].max()
    stats['IQR'] = stats['Q3'] - stats['Q1']
    stats['lim_inf'] = stats['Q1'] - 1.5 * stats['IQR']
    stats['lim_sup'] = stats['Q3'] + 1.5 * stats['IQR']

    # Líneas horizontales para los principales valores
    lineas = {
        'Media': ('mean', 'green', '--'),
        'Mediana': ('median', 'orange', '-.'),
        'Máximo': ('max', 'red', ':'),
        'Mínimo': ('min', 'blue', ':')
    }

    for etiqueta, (clave, color, estilo) in lineas.items():
        plt.axhline(y=stats[clave], color=color, linestyle=estilo,
                    label=f"{etiqueta}: {stats[clave]:.2f}")

    # Textos con valores de cuartiles y límites
    textos = {
        'Q1': stats['Q1'],
        'Q3': stats['Q3'],
        'Límite Inferior': stats['lim_inf'],
        'Límite Superior': stats['lim_sup']
    }

    for etiqueta, valor in textos.items():
        plt.text(0.85, valor, f"{etiqueta}: {valor:.2f}",
                 color='black', fontsize=9, va='center')

    # Ajustes finales
    plt.legend(loc='upper right', fontsize=8)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()

    # Retornar diccionario con estadísticas
    return stats
````

### Preparación del dataset para los ejercicios con groupby

#### pd.cut() divide una variable numérica continua (aquí df['quality']) en intervalos o “cajones”.

El argumento bins define los límites de esos intervalos:

1. (-∞, 5] → vinos de baja calidad

2. (5, 6] → vinos de calidad media

3. (6, ∞) → vinos de alta calidad

labels asigna nombres a cada intervalo (low, mid, high). El resultado es una serie categórica, que se guarda como una nueva columna df['quality_label']

#### Se clasifica el azúcar residual (residual sugar) en tres niveles:

1. (-∞, 2.0] → bajo (low)

2. (2.0, 4.0] → medio (mid)

3. (4.0, ∞) → alto (high)

El resultado se guarda en df['sugar_level'].

En resumen:

| Columna creada  | Variable base    | Rangos (bins)                | Etiquetas      | Objetivo                            |
| --------------- | ---------------- | ---------------------------- | -------------- | ----------------------------------- |
| `quality_label` | `quality`        | (-∞,5], (5,6], (6,∞)         | low, mid, high | Clasificar calidad del vino         |
| `sugar_level`   | `residual sugar` | (-∞,2.0], (2.0,4.0], (4.0,∞) | low, mid, high | Clasificar nivel de azúcar residual |


````python
# Preparación opcional de columnas auxiliares
if 'quality_label' not in df.columns:
    df['quality_label'] = pd.cut(
        df['quality'],
        bins=[-float('inf'), 5, 6, float('inf')],
        labels=['low','mid','high']
    )
if 'sugar_level' not in df.columns:
    sugar = df['residual sugar']
    df['sugar_level'] = pd.cut(
        sugar,
        bins=[-float('inf'), 2.0, 4.0, float('inf')],
        labels=['low','mid','high']
    )
df[['quality','quality_label','residual sugar','sugar_level']].head()
````

#### Explicación del código


`df.groupby('quality_label', observed=True)` Agrupa el DataFrame df según la columna quality_label.

Cada grupo corresponde a un valor distinto de 'quality_label' (por ejemplo, 'baja', 'media', 'alta').

El parámetro `observed=True` se usa cuando 'quality_label' es una columna de tipo categórico (Categorical). Hace que solo se incluyan las categorías que realmente aparecen en los datos. Mejora el rendimiento y evita mostrar categorías vacías.

`.agg(...)` El método agg() permite aplicar una o varias funciones de agregación sobre distintas columnas. Se usa la sintaxis nuevo_nombre = (columna, función). En este caso: `conteo=('quality', 'count')` cuenta cuántas filas hay por grupo de 'quality_label'.

`media_alcohol=('alcohol', 'mean')` calcula la media de 'alcohol' por grupo.

1. El resultado (`res`) es un nuevo DataFrame que muestra, para cada grupo:

El número de observaciones (conteo).

El promedio del alcohol (media_alcohol).

quality_label  .Por ejemplo, 'baja', 'media', 'alta'.

sugar_level    .Por ejemplo, 'bajo', 'medio', 'alto'.

Cada combinación de ambas columnas forma un grupo.

(baja, bajo)

(baja, medio)

(alta, medio)

Se utiliza cuando las columnas son de tipo categórico (Categorical).

Hace que solo se incluyan las combinaciones que realmente existen en los datos.

Selecciona la columna sulphates para realizar los cálculos.

quality_label	sugar_level	sulphates

**Ejercicio groupby 9.** Agrupa por `quality_label` y calcula los percentiles 25, 50 y 75 de `alcohol` en cada grupo. Devuelve un `DataFrame` con columnas `p25`, `p50`, `p75`.

Paso a paso
1️. `df.groupby('quality_label', observed=True)`

Agrupa el DataFrame df según los valores de la columna quality_label.

Cada grupo representa un nivel de calidad, por ejemplo: 'baja', 'media', 'alta'.

El parámetro `observed=True` se usa cuando la columna quality_label es categórica (Categorical).

Hace que solo se incluyan las categorías que realmente existen en los datos (evita filas vacías y mejora el rendimiento).

2️. `['alcohol']`

Selecciona la columna alcohol dentro de cada grupo.

Los cálculos se aplicarán únicamente sobre los valores de alcohol.

3. `.quantile([0.25, 0.5, 0.75])`

Calcula los cuantiles (también conocidos como percentiles) de la columna alcohol dentro de cada grupo.

Los valores usados son:

0.25, Primer cuartil (Q1) o percentil 25 (P25)

0.5, Mediana (Q2) o percentil 50 (P50)

0.75 , Tercer cuartil (Q3) o percentil 75 (P75)

Esto nos da una medida de cómo se distribuyen los valores de alcohol en cada categoría de calidad.

4. .unstack()

Convierte el índice jerárquico generado por groupby + quantile en columnas. Así, los distintos cuantiles (0.25, 0.5, 0.75) pasan a ser columnas en lugar de filas. Después de este paso, el resultado (q) tiene:

Filas = valores de quality_label

Columnas = los cuantiles calculados

5️. `q.columns = ['p25', 'p50', 'p75']`

Renombra las columnas para que sean más legibles:

p25: percentil 25

p50: percentil 50 (la mediana)

p75: percentil 75

6️. `res = q`

Guarda el resultado final en la variable res.

7. `res`

Muestra el DataFrame final con los tres cuantiles por categoría de calidad.

````python
# Escribe tu solución aquí
q = df.groupby('quality_label', observed=True)['alcohol'].quantile([0.25, 0.5, 0.75]).unstack()
q.columns = ['p25','p50','p75']
res = q
res
````

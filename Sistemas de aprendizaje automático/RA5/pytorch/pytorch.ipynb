## PyTorch

PyTorch es una herramienta fundamental en el campo del aprendizaje profundo; es decir, a las redes neuronales de gran envergadura o a las redes de perceptrones con múltiples capas. En su estructura más básica, estas redes consisten en una serie de niveles interconectados de manera secuencial. En este cuaderno de jupyter construiremos redes neuronales y diseñaremos modelos sencillos de aprendizaje profundo utilizando PyTorch.

###  Redes Neuronales en PyTorch
PyTorch ofrece una amplia gama de funcionalidades, pero su aplicación más frecuente es la creación de modelos de aprendizaje profundo.

La estructura más básica se puede diseñar empleando la clase Sequential, que consiste en una serie de capas interconectadas de manera lineal.

Es posible generar un modelo Sequential y establecer todas sus capas en un solo paso, dentro de paréntesis, siguiendo el orden de procesamiento desde la entrada hasta la salida.

1. Un modelo puede definirse como una secuencia de capas usando `Sequential`.

2. La primera capa debe tener el número de características de entrada.

3. El diseño del modelo (número de capas, neuronas) a menudo se determina por experimentación.

4. Se utilizan capas totalmente conectadas (`Linear`) en este ejemplo.

5. Se emplean funciones de activación después de cada capa: ReLU para las capas ocultas y sigmoide para la capa de salida ya que es un problema de clasificación binaria.

6. La estructura del modelo propuesto es:

   - Entrada: 8 variables

   - Primera capa oculta: 12 neuronas con ReLU

   - Segunda capa oculta: 8 neuronas con ReLU
   
   - Capa de salida: 1 neurona con sigmoide

Este enfoque permite crear un modelo de red neuronal para clasificación binaria, con una salida entre 0 y 1.

````python
import torch
import torch.nn as nn

modelo = nn.Sequential(
    # 8 entradas y 1ª capa oculta de 12 perceptrones
    nn.Linear(8, 12, bias = True, dtype=torch.float32), 
    # función de activación RelU perceptrones 1ª capa oculta
    nn.ReLU(),
    # 12 entradas (provienen de la 1ª capa oculta) y 8 perceptrones 2ª capa oculta 
    nn.Linear(12, 8, bias = True, dtype=torch.float32),
    # función de activación RelU perceptrones 2ª capa oculta
    nn.ReLU(),
    # 8 entradas (provienen de la 2ª capa oculta) y 1 perceptrones capa de salida 
    nn.Linear(8,1, bias = True, dtype=torch.float32),
    # función de activación para la capa de salida
    nn.Sigmoid()
)
````

## Capas, Activaciones y Propiedades de la Capa


PyTorch ofrece una amplia variedad de capas para redes neuronales, e incluso permite la creación de capas personalizadas. Algunas de las capas más frecuentes son:
- `nn.Linear(input, output)`: Capa densa o completamente conectada. Los dos argumentos representan las dimensiones de entrada y salida de esa capa.
- `nn.Conv2d(in_channel, out_channel, kernel_size)`: Capa de convolución 2D, utilizada en redes para procesamiento de imágenes.
- `nn.Dropout(probability)`: Capa de dropout, empleada para introducir regularización (técnica para reducir *overfitting*).
- `nn.Flatten()`: Transforma un tensor de entrada multidimensional en una dimensión.

Además de las capas, existen funciones de activación que se aplican elemento a elemento en un tensor. Estas funciones suelen utilizarse después de una capa y antes de la siguiente. Algunas funciones de activación comunes son:

- `nn.ReLU()`: Unidad lineal rectificada, la más utilizada actualmente.
- `nn.Sigmoid()` y `nn.Tanh()`: Funciones sigmoidea y tangente hiperbólica, populares en modelos más antiguos.
- `nn.Softmax()`: Se utiliza principalmente en la capa de salida de redes neuronales para problemas de clasificación multiclase,.

La mayoría de los componentes de un modelo aceptan dos argumentos adicionales: el dispositivo y el tipo de datos.

## Función de Pérdida y Optimizadores de Modelo

El entrenamiento de una red neuronal busca optimizar los pesos para generar la salida deseada. En el aprendizaje profundo, el método utilizado para optimizar estos pesos es el descenso de gradiente.

- Existen diversas variantes del descenso de gradiente. 
- Puedes seleccionar una preparando un optimizador para tu modelo. 
- Su uso implica definir una función de pérdida y minimizarla mediante el optimizador. 
- La función de pérdida indica cuán alejada está la salida del modelo de la salida deseada. 

### Función de pérdida (*Loss Functions*)

Las funciones de pérdida miden qué tan bien o mal está funcionando el modelo.  El objetivo del entrenamiento es minimizar esta pérdida.

En PyTorch, puedes manipular directamente el tensor de salida del modelo para calcular la pérdida. Sin embargo, también puedes utilizar las funciones proporcionadas por PyTorch para el cálculo de esta pérdida. Por ejemplo:

```python
loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(output, label)
```

En este caso, `loss_fn` es una función, y `loss` es un tensor que admite diferenciación automática. La diferenciación se inicia con `loss.backward()`.

Algunas funciones de pérdida en PyTorch son:

- `nn.MSELoss()`: Error cuadrático medio, útil en problemas de regresión.
- `nn.CrossEntropyLoss()`: Pérdida de entropía cruzada, útil en problemas de clasificación.
- `nn.BCELoss()`: Pérdida de entropía cruzada binaria, útil en problemas de clasificación binaria.

## 1.  Error cuadrático medio. *Mean Squared Error* (MSE)

$$
\mathcal{L}_{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$


- $y_i$: valor real
- $\hat{y}_i$: predicción del modelo
- Penaliza más los errores grandes porque están al cuadrado

### Uso en PyTorch

- Regresión
- No es la función habitual para clasificación

```python
import torch.nn as nn

loss_fn = nn.MSELoss()
```

## 2. Pérdida de entropía cruzada. *Binary Cross Entropy* (BCE)

$$
\mathcal{L}_{BCE} = - \frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{y}_i) + (1 - y_i)\log(1 - \hat{y}_i) \right]
$$


- $y_i \in \{0,1\}$: etiqueta real
- $\hat{y}_i \in (0,1)$: salida del modelo tras aplicar Sigmoid


Si $y = 1$:

$$
\mathcal{L} = -\log(\hat{y})
$$

Si $y = 0$:

$$
\mathcal{L} = -\log(1-\hat{y})
$$

### Uso en Pytorch

- Clasificación binaria
- Se suele usar cuando el modelo termina con una capa `Sigmoid()`

```python
import torch.nn as nn

nn.Sigmoid()
loss_fn = nn.BCELoss()

## 3. Binary Cross Entropy with Logits (*BCEWithLogitsLoss*)

La última capa lineal del modelo produce un valor real (logit):

$$
z = Wx + b
$$

Después, este valor se puede transformar en probabilidad mediante la función Sigmoid:

$$
\hat{y} = \sigma(z) = \frac{1}{1 + e^{-z}}
$$


La pérdida asociada es:

$$
\mathcal{L} = - \left[ y \log(\sigma(z)) + (1-y)\log(1-\sigma(z)) \right]
$$


- $y \in \{0,1\}$: valor real  
- $z$: salida de la red antes de aplicar Sigmoid (logit)  
- $\sigma(z)$: probabilidad predicha  

`BCEWithLogitsLoss` combina en una sola función:

- la activación Sigmoid  
- la pérdida Binary Cross Entropy  

Esto mejora la estabilidad numérica y evita problemas de overflow, underflow, y gradientes inestables  

### Uso en PyTorch

```python
import torch.nn as nn

loss_fn = nn.BCEWithLogitsLoss()

## 4. Cross Entropy (Clasificación Multiclase)


$$
\mathcal{L} = - \sum_{i=1}^{C} y_i \log(\hat{y}_i)
$$

donde las probabilidades de salida se obtienen mediante Softmax:

$$
\hat{y}_i = \frac{e^{z_i}}{\sum_{j=1}^{C} e^{z_j}}
$$


- $C$: número de clases
- $y_i$: valor real asociado a la clase $i$
- $\hat{y}_i$: probabilidad predicha para la clase $i$

Con esta función de pérdida, si el modelo predice una probabilidad alta para la clase correcta, la pérdida será pequeña. Si predice una probabilidad baja para la clase correcta, la pérdida será grande.

### Uso en PyTorch

Se utiliza en problemas de **clasificación multiclase**, por ejemplo:

- reconocimiento de dígitos
- clasificación de imágenes
- clasificación de texto en varias categorías

```python
import torch.nn as nn

loss_fn = nn.CrossEntropyLoss()
```

## Proceso de entrenamiento de una red neuronal

Una vez definidos el modelo, la función de pérdida y el optimizador, el modelo está listo para ser entrenado con datos.

### Épocas y lotes

El entrenamiento de una red neuronal se realiza generalmente en épocas y lotes:

- **Época (epoch):**  
  Consiste en recorrer todo el conjunto de datos por el modelo una vez.

- **Lote (batch):**  
  Es un subconjunto de muestras utilizado en una iteración del algoritmo de descenso de gradiente dentro de cada época.

### Proceso de entrenamiento

El proceso se repite durante múltiples épocas hasta que el modelo alcanza un rendimiento satisfactorio.

### Parámetros importantes

- **Tamaño del lote (batch size):**
  - Está limitado por la memoria del sistema
  - Influye en la cantidad de cómputo necesario
  - Afecta la estabilidad del entrenamiento
  - 

- **Número de épocas (epochs):**
  - Determina cuántas veces el modelo ve el conjunto de datos completo
  - Más épocas, generalmente, implica mejor aprendizaje

### Ajuste de hiperparámetros

El número de épocas y el tamaño del lote se ajustan normalmente mediante:

- prueba y error  
- experimentación  
- validación del modelo

````python
n_epocas = 10
tamanho_batch = 10

for epoca in range(n_epocas):
    for i in range(0, len(X_tensor), tamanho_batch):

        # Seleccionamos un subconjunto de datos (batch)
        X_batch = X_tensor[i : i + tamanho_batch ]

        # El modelo hace una predicción (forward pass)
        y_pred  = modelo(X_batch)

        # Seleccionamos las etiquetas reales correspondientes al batch
        y_batch = y_tensor[i : i + tamanho_batch]

        # Calculamos el error entre predicción y valor real
        loss    = loss_fn(y_pred, y_batch)

        # PyTorch acumula gradientes, no los reemplaza. Borra gradientes anteriores
        optimizador.zero_grad()

        # # Backpropagation: Cálculo del gradiente. Matemáticamente: dL/dW
        loss.backward()

        # # Actualizamos los pesos del modelo
        # W = W - learning_rate * gradiente(derivadas)
        optimizador.step()

   
    # loss.item() muestra el tensor como un número. Es decir, extrae el valor numérico de un tensor de un solo elemento.  
    print(f'Epoca :{epoca: 3d}, pérdida (loss) : {loss.item(): .3f}')       
        
````

## Relación entre la función de pérdida y el código en PyTorch

Durante el entrenamiento de una red neuronal, el objetivo es minimizar la función de pérdida:

$$
\mathcal{L}_{BCE}
$$

Esto se realiza mediante el cálculo de gradientes y la actualización de los pesos:

$$
W = W - \eta \cdot \nabla \mathcal{L}_{BCE}
$$


### 1. Cálculo de la pérdida

```python
loss = loss_fn(y_pred, y_batch)
```

$$
\mathcal{L}_{BCE} = - \frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{y}_i) + (1 - y_i)\log(1 - \hat{y}_i) \right]
$$


### 2. Cálculo de gradientes (Backpropagation)

Durante este paso, PyTorch calcula automáticamente los gradientes de la función de pérdida respecto a todos los pesos del modelo.

```python
loss.backward()
```

### Gradiente de la función de pérdida

El gradiente de la función de pérdida respecto a los pesos se define como:

$$
\nabla \mathcal{L}_{BCE} = \frac{\partial \mathcal{L}_{BCE}}{\partial W}
$$

### Significado

- $\nabla \mathcal{L}_{BCE}$ indica cómo cambia la función de pérdida cuando cambian los pesos $W$
- Mide la sensibilidad del error respecto a los parámetros del modelo


- Si $\nabla \mathcal{L}_{BCE} > 0$, el error aumenta y los pesos deben disminuir  
- Si $\nabla \mathcal{L}_{BCE} < 0$, el error disminuye y los pesos deben aumentar  

El gradiente proporciona la dirección en la que el error crece más rápidamente.  
Para minimizar la función de pérdida, los pesos se actualizan en la dirección opuesta al gradiente.

### 3. Actualización de pesos

Durante este paso, los pesos del modelo se actualizan para reducir la función de pérdida.

```python
optimizador.step()
```

$$
W = W - \eta \cdot \nabla L_{BCE}
$$

### 4. Reinicio de gradientes

Antes de calcular nuevos gradientes, es necesario reiniciar los anteriores:

```python
optimizador.zero_grad()
```

Cada actualización de pesos debe basarse solo en el batch actual, no en acumulación de información anterior.

````python
import torch
import torch.nn as nn

modelo = nn.Sequential(
    # 8 entradas y 1ª capa oculta de 12 perceptrones
    nn.Linear(8, 12, bias = True, dtype=torch.float32), 
    # función de activación RelU perceptrones 1ª capa oculta
    nn.ReLU(),
    # 12 entradas (provienen de la 1ª capa oculta) y 8 perceptrones 2ª capa oculta 
    nn.Linear(12, 8, bias = True, dtype=torch.float32),
    # función de activación RelU perceptrones 2ª capa oculta
    nn.ReLU(),
    # 8 entradas (provienen de la 2ª capa oculta) y 1 perceptrones capa de salida 
    nn.Linear(8,1, bias = True, dtype=torch.float32),
    # función de activación para la capa de salida
    nn.Sigmoid()
)
````

````python
# definimos la función de pérdida
loss_fn = nn.BCELoss()
import torch.optim as optim
# definimos el optimizador
optimizador = optim.Adam(modelo.parameters(),
                        lr = 0.001)
````

````python
# número de recorridos del dataset completo
n_epocas = 50

# número de subconjuntos en los que se divide el dataset (batches)
tamanho_batch = 10

# número de batches (enteros) que hay en una época. Ignora los datos sobrantes
batches_por_epoca = len(X_train) // tamanho_batch  

# Listas para guardar la evolución de las métricas
lista_loss_train = []
lista_acc_train = []
lista_acc_test = []

# Repite todo el entrenamiento n_epocas veces
for epoca in range(n_epocas):

    # MODELO EN MODO ENTRENAMIENTO
    modelo.train()

    # acumuladores para calcular métricas medias
    loss_acumulada = 0.0
    acc_acumulada = 0.0

    # tqdm.trange(...) crea una barra de progreso y as bar guarda ese objeto en la variable bar
    # bar = barra de progreso + iterador (representa/itera sobre los batches).
    # unit = 'batch' para que aprezca en la etiqueta cuántos batchs se procesan por segundo
    with tqdm.trange(batches_por_epoca, unit = 'batch') as bar:
    
        bar.set_description(f'Época {epoca}')

        # Recorremos todos los batches de esta época. i es el número de batch
        for i in bar:
         
            # Calculamos el índice inicial del batch actual
            # Si i = 0, empieza en 0
            # Si i = 1, empieza en 10
            # Si i = 2, empieza en 20
            inicio_batch = i * tamanho_batch

            # Calculamos el índice final del batch actual
            fin_batch = inicio_batch + tamanho_batch
    
            # Seleccionamos un subconjunto de datos (batch)
            X_batch = X_train[inicio_batch : fin_batch ]

            # Seleccionamos las etiquetas reales correspondientes al batch
            y_batch = y_train[inicio_batch : fin_batch]
    
            # El modelo hace una predicción (forward pass)
            y_pred  = modelo(X_batch)
    
            # Calculamos el error entre predicción y valor real
            loss    = loss_fn(y_pred, y_batch)
    
            # PyTorch acumula gradientes, no los reemplaza. Borra gradientes anteriores
            optimizador.zero_grad()
    
            # Backpropagation: Cálculo del gradiente. Matemáticamente: dL/dW
            # calculamos automáticamente las derivadas de la pérdida respecto a los parámetros del modelo
            loss.backward()
    
            # Actualizamos los pesos del modelo  W = W - learning_rate * gradiente(derivadas)
            optimizador.step()

            acc = (y_pred.round() == y_batch).float().mean()

            loss_acumulada += loss.item()
            acc_acumulada += acc.item()

            # Mostramos el valor actual de la pérdida en la barra de progreso
            # La barra es de la época, pero el contenido (loss, acc) es del último batch procesado
            # set_postfix muestra las métricas del batch actual en tiempo real (la salida muestra la info del último batch de la época)
            bar.set_postfix(loss = f'{float(loss): .4f}',
                           acc = f'{float(acc) * 100:.4f}%')

    # métricas medias de entrenamiento de la época
    loss_media = loss_acumulada / batches_por_epoca
    acc_media = acc_acumulada / batches_por_epoca
            
    # MODELO EN MODO EVALUACIÓN
    
    # Cambia el modelo a modo evaluación (inferencia) con el conjunto de test. Durante el entrenamiento, el modelo aprende ajustando sus pesos.
    # Ahora ya no queremos aprender, solo queremos hacer predicciones sobre X_test, compararlas con las etiquetas reales y_test
    # y calcular la exactitud (accuracy)
    modelo.eval()

    # No calculamos gradientes, más rápido y menos memoria.
    with torch.no_grad():
        y_pred_test = modelo(X_test)
        acc_test = (y_pred_test.round() == y_test).float().mean()

    # Guardamos métricas de la época
    lista_loss_train.append(loss_media)
    lista_acc_train.append(acc_media)
    lista_acc_test.append(acc_test.item())

    # mostramos métricas del modelo para el conjunto de test
    print(
        f"Época {epoca}: "
        f"loss_train={loss_media:.4f}, "
        f"acc_train={acc_media:.4f}, "
        f"acc_test={acc_test:.4f}"
    )       
    
  
````

## Aplicaciones y Conceptos de Árboles de Decisión

Un Árbol de Decisión es un modelo predictivo que divide los datos en ramas para llegar a conclusiones o predicciones, similares a cómo uno tomaría decisiones en la vida real, sopesando diferentes opciones y sus consecuencias.

#### 1. Árboles de Decisión
Estructuras de modelado que representan decisiones y sus posibles consecuencias, utilizadas para simplificar la toma de decisiones complejas.

#### 2. Análisis de Decisiones
Facilitan la visualización y evaluación de múltiples opciones y resultados, ayudando a elegir el curso de acción más beneficioso.

#### 3. Minería de Datos
Describen y predicen datos, identificando patrones y permitiendo la toma de decisiones informadas basadas en *insights* extraídos de grandes conjuntos de datos.

#### 4. Aplicabilidad y Popularidad
Su uso extendido en estadística, minería de datos y aprendizaje automático se debe a su capacidad para construir modelos predictivos de manera intuitiva y fácilmente interpretable.

### Funcionamiento de los Árboles de Decisión

#### 1. Estructura del Árbol
Comienza con un nodo raíz que se ramifica en nodos internos representando preguntas y termina en hojas que simbolizan los resultados finales.

#### 2. Proceso de Segmentación
Los algoritmos dividen los datos en subconjuntos a través de preguntas cuantitativas, buscando maximizar la homogeneidad dentro de los subconjuntos.

#### 3. Algoritmos Comunes
Incluyen ID3, C4.5, CART y CHAID, cada uno con estrategias específicas para la selección de las mejores preguntas y divisiones.

#### 4. Predicciones y Clasificaciones
Resultan en modelos que pueden predecir resultados o clasificar datos basándose en un conjunto de decisiones lógicas derivadas del árbol.

````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs # Importamos la función para generar datos artificiales
import pandas as pd
import matplotlib.pyplot as plt

# Generamos 3 grupos de 40 puntos
X, y = make_blobs(
    n_samples=[40, 40, 40],
    centers=[(2, 2), (7, 3), (6, 9)],
    random_state=42
)

# Convertimos a DataFrame para visualizarlo mejor
df = pd.DataFrame(X, columns=["x1","x2"])
df["y"] = y

print(df)
print(df["y"].value_counts().sort_index())

# Representación
plt.scatter(df["x1"], df["x2"], c=df["y"])
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Tres grupos globulares de 40 puntos")
plt.grid(True)
plt.show()
    
````

````python
# Importamos el clasificador árbol de decisión
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text

print("\nNúmero de ejemplos por clase:")
print(df["y"].value_counts().sort_index())

# Creamos el clasificador CART usando Gini
# max_depth se deja suficientemente grande para permitir crecer el árbol
clf = DecisionTreeClassifier(
    criterion="gini",
    splitter="best",
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42
)
# Entrenamos el modelo
clf.fit(X, y)

# Dibujamos el árbol
plt.figure(figsize=(15, 10))

plot_tree(
    clf,
    feature_names=["x1", "x2"],
    class_names=["clase 0", "clase 1", "clase 2"],
    filled=True,
    rounded=True,               # bordes redondeados de los nodos del árbol
    impurity=True,              # valor de Giny Impurity
    proportion=False,
    fontsize=10
)

plt.title("Árbol de decisión aprendido por scikit-learn")
plt.show()


# Dibujamos las regiones de decisión
# Calculamos límites del plano
x1_min = X[:, 0].min() - 0.5 # X[:, 0] todos los valores de x1
x1_max = X[:, 0].max() + 0.5
x2_min = X[:, 1].min() - 0.5 # X[:, 1] todos los valores de x2
x2_max = X[:, 1].max() + 0.5

# Creamos una malla de puntos.Crea una rejilla 2D de puntos.
# xx valores de x1 repetidos en filas e yy valores de x2 repetidos en columnas
xx, yy = np.meshgrid(    
    np.linspace(x1_min, x1_max, 500),
    np.linspace(x2_min, x2_max, 500)
)
print(f'xx:\n {xx}')
print(f'type(xx): {type(xx)}')
print(f'dimensiones(xx): {(xx.shape)}')
print(f'xx.ravel(): {xx.ravel()}')
# Predecimos la clase en cada punto de la malla
print(f'np.c_[xx.ravel(), yy.ravel()]: {np.c_[xx.ravel(), yy.ravel()]}')
Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
print(f'Z = {Z}')

Z = Z.reshape(xx.shape) # Volver a forma de rejilla

print(f'Z = {Z}')
# Dibujamos las regiones
plt.figure(figsize=(10, 8))

# Esta función dibuja regiones coloreadas en el plano según los valores de Z
# contourf convierte números en zonas de color. Dibuja regiones coloreadas según la clase
plt.contourf(xx, yy, Z,
             alpha=0.3, # transparencia
             cmap="Set3") # mapa de color

# Dibujamos la frontera entre la regiones de decisión
plt.contour(xx, yy, Z,
            colors="r", # color del límite de decisión
            linewidths=1) # ancho del límite de decisión 

# Dibujamos los puntos reales
plt.scatter(df["x1"], df["x2"], c=df["y"], edgecolor="k")

plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Regiones de decisión del árbol")
plt.grid(True)
plt.show()


# Información interna del árbol

# Mostramos número de nodos, profundidad y número de hojas
print("\nNúmero total de nodos del árbol:", clf.tree_.node_count)
print("Profundidad del árbol:", clf.tree_.max_depth)
print("Número de hojas:", clf.get_n_leaves())
````

# Limitaciones de los árboles de decisión

- La versión implementada en Scikit-learn puede producir árboles de decisión complejos que producen **overfitting** sobre el conjunto de datos.

- La implementación del algoritmo se basa en la obtención de **óptimos locales** entre el nodo padre y el nodo hijo. Como consecuencia, no hay manera de asegurar que el árbol alcance una forma óptima global.

- Los árboles de decisión pueden ser **inestables**. Una pequeña variación en el conjunto de datos puede producir que se genere un árbol completamente diferente.

- Pueden mitigarse gran parte de los efectos de estas limitaciones si se utilizan conjuntos aleatorios de árboles que toman decisiones conjuntas (**Random Forest**).

# Overfitting

El 'overfitting' o sobreajuste ocurre cuando nuestro árbol se adapta demasiado a los datos de entrenamiento, capturando incluso el ruido o las anomalías, lo cual afecta negativamente su capacidad de generalizar y realizar predicciones precisas en datos no vistos anteriormente.
En este caso, existe un número elevado de características o de valores atípicos y, entonces, es posible que el árbol de decisión genere un número muy elevado de ramas, lo que puede provocar problemas de generalización.

Una forma de reducir este problema es aplicar un *criterio de parada* mediante hiperparámetros del modelo.

## Hiperparámetros para controlar el crecimiento del árbol

### `max_depth`
Determina la profundidad máxima del árbol.

```python
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier(max_depth=3)

# Si este valor es pequeño, el árbol será más simple y tendrá menos riesgo de sobreajuste.
# Si es muy grande o None, el árbol puede crecer demasiado y ajustarse en exceso a los datos de entrenamiento

### `min_samples_split`

Determina el número mínimo de muestras que debe tener un nodo para poder dividirse.

```python
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier(min_samples_split=10)

# Si un nodo tiene menos muestras que este valor, ya no se divide y se convierte en hoja.

### `min_samples_leaf`

Determina el número mínimo de muestras que debe tener una hoja.

```python
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier(min_samples_leaf=5)

# Este parámetro evita que se creen hojas con muy pocas muestras, que suelen ser poco representativas y favorecen el overfitting.


### `max_leaf_nodes`

Determina el número máximo de hojas que puede tener el árbol.

```python
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier(max_leaf_nodes=8)

# Si se limita el número de hojas, el árbol no podrá crecer indefinidamente y el modelo será más simple.

### Conjunto de datos: Detección de malware en Android
Descripción
The sophisticated and advanced Android malware is able to identify the presence of the emulator used by the malware analyst and in response, alter its behavior to evade detection. To overcome this issue, we installed the Android applications on the real device and captured its network traffic. See our publicly available Android Sandbox.

CICAAGM dataset is captured by installing the Android apps on the real smartphones semi-automated. The dataset is generated from 1900 applications with the following three categories:

1. Adware (250 apps)

Airpush: Designed to deliver unsolicited advertisements to the user’s systems for information stealing.

Dowgin: Designed as an advertisement library that can also steal the user’s information.

Kemoge: Designed to take over a user’s Android device. This adware is a hybrid of botnet and disguises itself as popular apps via repackaging.

Mobidash: Designed to display ads and to compromise user’s personal information.

Shuanet: Similar to Kemoge, Shuanet also is designed to take over a user’s device.

2. General Malware (150 apps)

AVpass: Designed to be distributed in the guise of a Clock app.

FakeAV: Designed as a scam that tricks user to purchase a full version of the software in order to re-mediate non-existing infections.

FakeFlash/FakePlayer: Designed as a fake Flash app in order to direct users to a website (after successfully installed).

GGtracker: Designed for SMS fraud (sends SMS messages to a premium-rate number) and information stealing.

Penetho: Designed as a fake service (hacktool for Android devices that can be used to crack the WiFi password). The malware is also able to infect the user’s computer via infected email attachment, fake updates, external media and infected documents.

3. Benign (1500 apps)

2015 GooglePlay market (top free popular and top free new)

2016 GooglePlay market (top free popular and top free new)

## CICFlowMeter (Sonda de Red)

Es una herramienta de ciberseguridad que permite extraer características (*features*) de flujos de red a partir de archivos PCAP, generando datasets listos para Machine Learning.

https://www.youtube.com/watch?v=EaYyYy9uvt0


### ¿Qué hace?
- Convierte paquetes de red en **flujos (flows)**
- Calcula **más de 80 características** por flujo
- Exporta los datos en formato **CSV**

### ¿Qué es un flow?
Un flujo agrupa paquetes que comparten:
- IP origen y destino
- Puertos origen y destino
- Protocolo (TCP/UDP)

### Características generadas
- Duración del flujo  
- Número de paquetes y bytes  
- Tiempos entre paquetes  
- Flags TCP (SYN, ACK, etc.)  
- Estadísticas (media, varianza, desviación)

### Aplicaciones
- Detección de intrusiones (IDS)
- Análisis de tráfico de red
- Creación de datasets para Machine Learning

### Flujo de trabajo típico

PCAP → CICFlowMeter → CSV → Modelo ML → Evaluación

### Relación con Machine Learning
Permite aplicar directamente algoritmos como:
- Decision Trees  
- Random Forest  
- Cross-validation

````python
df = pd.read_csv('TotalFeatures-ISCXFlowMeter.csv')
df.head()
````

````python
df.info()
````

````python
df['class'].value_counts()
````

````python
df['class'] = df['class'].map({
    'benign': 0,
    'asware': 1,
    'GeneralMalware': 2
})
````

````python
df['class'].value_counts()
````

````python
X = df.drop(columns=['class'])  # eliminamos la columna 'class', nos quedamos con el conjunto de variables predictoras
y = df['class']                 # variable a predecir
````

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,        # 20% para test, 80% para entrenamiento
    random_state=42,      # resultados reproducibles
    stratify=y            # mantiene proporción de clases. Muy importante en este caso
)

y_test.value_counts()
````

````python
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier(
    max_depth=10,             # el árbol puede crecer sin límite predefinido de niveles. Colocamos 10 niveles
    class_weight="balanced",
    random_state=42
)

clf.fit(X_train, y_train)
````

````python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Predicciones sobre el conjunto de test
y_pred = clf.predict(X_test)

# Accuracy
acc = accuracy_score(y_test, y_pred)
print("Accuracy:", acc)

# Informe de clasificación
print("\nClassification report:")
print(classification_report(y_test,
                            y_pred,
                            target_names=["benign", "asware", "GeneralMalware"]))
````

## Evaluación del modelo: DecisionTreeClassifier

### Accuracy

$$
\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}
$$

$$
\text{Accuracy} = 0.7596
$$

El modelo clasifica correctamente aproximadamente el **76%** de las muestras.

En datasets desbalanceados, la accuracy no es sufciente para interpretar la bondad del modelo, ya que está dominada por la clase mayoritaria (*benign*).

## Definición de métricas

### Precision

$$
\text{Precision} = \frac{TP}{TP + FP}
$$

De todas las muestras que el modelo predice como una clase, ¿cuántas son correctas?


### Recall

$$
\text{Recall} = \frac{TP}{TP + FN}
$$

De todas las muestras reales de una clase, ¿cuántas detecta el modelo?

### F1-score

$$
F1 = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}
$$

Media armónica entre precision y recall.

## Interpretación por clase

### Clase benign

- Precision = 0.96 (muy pocos **FP**)
- Recall = 0.73  (existen **FN**

El modelo es preciso pero no detecta todos los **benign**.

### Clase asware

- Precision = 0.61  (hay bastantes **FP**)
- Recall = 0.86  (pocos **FN**)

El modelo detecta bien asware, pero comete errores clasificando otras clases como asware.


### Clase GeneralMalware

- Precision = 0.07  (muchísimos **FP**)
- Recall = 0.78  (pocos **FN**)

El modelo detecta malware (alto recall), pero genera muchísimos falsos positivos y predice malware cuando no lo es.

## Métricas globales

### Macro average

$$
\text{Macro Avg} = \frac{1}{N} \sum \text{métrica por clase}
$$

- Precision = 0.55  
- Recall = 0.79  
- F1-score = 0.55  

Trata todas las clases por igual, refleja mejor el problema del desbalanceo.


### Weighted average

$$
\text{Weighted Avg} = \sum \left( \text{métrica}_i \cdot \frac{\text{support}_i}{\text{total}} \right)
$$

- Precision = 0.87  
- Recall = 0.76  
- F1-score = 0.79  

Está dominado por la clase *benign* (mayoritaria).


## Conclusión

- El modelo tiene buena accuracy, pero depende del desbalanceo del dataset.
- Funciona bien en clases mayoritarias.
- Falla gravemente en GeneralMalware, precision muy baja. Exceso de FP.
- No es adecuado en su estado actual para detección real de malware.

## Cálculo de métricas

### Precision
De todas las muestras que el modelo predice como de clase GeneralMalware, ¿cuántas son realmente de tipo GeneralMalware?

$$
\text{Precision} = \frac{TP}{TP + FP}
= \frac{743}{743 + 10660}
= \frac{743}{11403}
\approx 0.065 \approx 0.07
$$

### Recall
De todas las muestras que son realmente de clase GeneralMalware, ¿cuántas detecta el modelo como GeneralMalware?

$$
\text{Recall} = \frac{TP}{TP + FN}
= \frac{743}{743 + 206}
= \frac{743}{949}
\approx 0.783 \approx 0.78
$$

### F1-score


$$
F1 \approx 0.12
$$

Detecta bastante malware, recall alto (0.78). Pero:

$$
FP = 10660 \gg TP = 743
$$

Genera MUCHÍSIMOS falsos positivos. Es decir, de todos los casos que el modelo dice “esto es malware”:

$$
TP + FP = 11403
$$

Correctos 743 e incorrectos 10660  

### Porcentaje real

$$
\text{Precision} \approx 7\%
$$

De cada 100 alertas de malware sólo 7 son reales 93 son falsas alarmas.

## Conclusión

El modelo tiene un comportamiento muy agresivo para detectar malware: consigue un recall alto, pero a costa de generar una cantidad masiva de falsos positivos (FP). ### Análisis de las regiones de decisión


Es un plano con dos variables. En el eje X  `min_flowpktl` y, en el eje Y, `Init_Win_bytes_forward`. Los puntos negros son datos reales. El color de fondo es lo que predice el árbol.

El árbol divide el espacio con reglas del tipo:

$$
x_j \leq t
$$

Esto significa que hace cortes **verticales** (sobre X) y cortes **horizontales** (sobre Y). Por lo tanto, se crean **rectángulos**  y cada rectángulo es una una clase.


#### Zona derecha (valores altos de x)

A partir de cierto valor de `min_flowpktl` todo el fondo tiene el **mismo color**. El árbol ha aprendido que:

$$
\text{si } min\_flowpktl \text{ es alto, los puntos están en la misma clase}
$$

Es una zona simple, fácil de clasificar y se cometen pocos errores.


#### Zona izquierda (valores bajos de x)

Aquí hay muchas franjas verticales, regiones pequeñas y cambios constantes de color. Por lo que el árbol **no sabe separar bien las clases**. Cuando el modelo no separa bien hace muchas divisiones y crea reglas muy específicas. Esto suele indicar Overfitting local.

Muchos puntos están mezclados en la zona izquierda y no hay separación clara entre clases. Entonces el árbol intenta “forzar” la separación pero no puede hacerlo bien.

#### Consecuencia en métricas

GeneralMalware tiene precision muy baja (~0.07). El modelo crea pequeñas zonas donde predice “malware” pero dentro hay muchos puntos que NO son malware. En este caso, hay pocos TP. Es decir, acierta algunos malware y muchos FP (confunde otras clases).

$$
FP \gg TP
$$

precision baja y F1 bajo. El problema no es solo el modelo, las variables no separan bien las clases.

El árbol solo puede hacer cortes verticales y horizontales. No puede hacer fronteras diagonales o separaciones complejas


#### Resumen

- El árbol depende mucho de `min_flowpktl`. Es decir, en valores altos hay buena separación y en valores bajos hay fuerte solapamiento de clases
- El modelo crea muchas regiones pequeñas, posible overfitting
- Esto provoca muchos falsos positivos, especialmente en la clase minoritaria (GeneralMalware)

#### Conclusión

Las regiones de decisión muestran que el modelo no puede separar correctamente las clases con estas dos variables. En zonas donde los datos están mezclados, el árbol genera muchas subdivisiones y comete errores, lo que explica el bajo rendimiento en precisión para la clase GeneralMalware.


# Random Forests

- Se corresponde con un *Ensemble* de Árboles de decisión

- Se entrena utilizando la técnica de *Ensemble Learning* conocida como **Bagging**

- En lugar de buscar el mejor par *(característica, valor)* de entre todas las características del conjunto de datos, lo hace respecto a un **subconjunto aleatorio de características**

- Los hiperparámetros más relevantes se corresponden con:

  - **n_estimators**: Número de árboles de decisión en el *Ensemble*

  - **max_depth**: Profundidad máxima de los árboles de decisión que componen el *Ensemble*

- Permiten **medir la importancia relativa de cada característica** del conjunto de datos en la realización de la predicción

````python
# Importamos la clase RandomForestClassifier desde sklearn
from sklearn.ensemble import RandomForestClassifier

# Creamos una instancia del modelo Random Forest
rf = RandomForestClassifier(
    n_estimators=100,   # Número de árboles que compondrán el bosque. Más árboles mejor generalización, pero mayor coste computacional.

    max_depth=10,  # Profundidad máxima de cada árbol.
    # None significa que los árboles crecerán hasta que las hojas sean puras o no se puedan dividir más.
    # Puede provocar overfitting si no se controla.

    max_features="sqrt", # Número de variables consideradas en cada división del árbol.
                         # 'sqrt' raíz cuadrada del número total de variables (valor típico en clasificación).

    random_state=42, # Fija la semilla aleatoria para que los resultados sean reproducibles.
    n_jobs=-1   # Número de CPUs utilizadas en paralelo. -1 usa todos los cores disponibles.
)

# Entrenamos el modelo con los datos de entrenamiento
rf.fit(X_train, y_train) # El modelo aprende múltiples árboles a partir de subconjuntos aleatorios de los datos (Bagging)

# Realizamos predicciones sobre el conjunto de test
y_pred = rf.predict(X_test) # Cada árbol vota una clase y el modelo devuelve la clase mayoritaria (votación)
````

````python
# EVALUACIÓN DEL MODELO
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# accuracy (Exactitud). 
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy) # Accuracy = (Número de predicciones correctas) / (Número total de predicciones)

# Matriz de confusión
cm = confusion_matrix(y_test, y_pred)

# Generamos un informe completo con métricas por clase
report = classification_report(y_test, y_pred)

# Mostramos el informe
print("\nClassification Report:")
print(report)

# Este informe incluye:
#
# - Precision:
#   De todas las predicciones positivas, cuántas son correctas.  Precision = TP / (TP + FP)
#
# - Recall (Sensibilidad):
#   De todos los positivos reales, cuántos detecta el modelo.   Recall = TP / (TP + FN)
#
# - F1-score:
#   Media armónica entre precision y recall.  F1 = 2 * (Precision * Recall) / (Precision + Recall)
#
# - Support:
#   Número de muestras reales de cada clase

# Extraemos la importancia de cada característica
importances = rf.feature_importances_

# Mostramos las importancias
print("\nImportancia de las variables:")
for i, imp in enumerate(importances):
    print(f"Feature {i}: {imp:.4f}")


# Indica cuánto contribuye cada variable a la decisión del modelo.
# Cuanto mayor es el valor, más importante es la variable.
````

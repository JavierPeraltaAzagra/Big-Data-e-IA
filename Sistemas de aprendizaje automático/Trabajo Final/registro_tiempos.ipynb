```python
import pandas as pd
import matplotlib.pyplot as plt

# CPUs - 4a
real = [528.44, 504.90, 525.45, 573.11, 699.77]
user = [1029.00, 981.89, 2042.66, 4465.47, 21543.49]
sys = [8.17, 8.40, 10.03, 10.41, 17.12]
cpus = [1, 2, 4, 8, 16]

cpu_df_4a = pd.DataFrame({"CPUs": cpus, "real": real, "user": user, "sys": sys})
cpu_df_4a

plt.figure(figsize=(7, 4))
plt.plot(cpus, real, marker="o", label="real")
plt.plot(cpus, user, marker="o", label="user")
plt.plot(cpus, sys, marker="o", label="sys")
plt.title("Tiempos vs numero de CPUs")
plt.xlabel("Numero de CPUs")
plt.ylabel("Tiempo")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 4b
loss_train = [0.2475, 0.2466, 0.2451, 0.2436, 0.2453]
acc_train = [0.8814, 0.8819, 0.8843, 0.8863, 0.8844]
acc_test = [0.8806, 0.8795, 0.8859, 0.8861, 0.8876]
cpus = [1, 2, 4, 8, 16]

cpu_df_4b = pd.DataFrame({"CPUs": cpus, "loss_train": loss_train, "acc_train": acc_train, "acc_test": acc_test})
cpu_df_4b

plt.figure(figsize=(7, 4))
plt.plot(cpus, loss_train, marker="o", label="loss_train")
plt.plot(cpus, acc_train, marker="o", label="acc_train")
plt.plot(cpus, acc_test, marker="o", label="acc_test")
plt.title("Tiempos vs numero de CPUs")
plt.xlabel("Numero de CPUs")
plt.ylabel("Tiempo")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

path = "TotalFeatures-ISCXFlowMeter.csv"

df = pd.read_csv(path)
print(df.shape)
print(df['class'].value_counts(dropna=False))

mapping = {'benign': 0, 'adware': 1, 'GeneralMalware': 1}
unmapped = df.loc[~df['class'].isin(mapping.keys()), 'class'].unique()
print(unmapped)

df['class'] = df['class'].map(mapping)

if df['class'].isna().any():
    raise ValueError("Hay clases no mapeadas en 'class'")

df['class'] = df['class'].astype('int64')
```
